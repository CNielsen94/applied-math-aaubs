### The exponential function

$f(x) = b^x$

An exponential function is a mathematical function in which a variable appears in the exponent. The most basic form of an exponential function is $y = b^x$, where b is a positive constant different from 1. If b is greater than 1, the function grows exponentially, while if b is between 0 and 1, the function decays exponentially.

Exponential functions are very common in many scientific, technical, and economic contexts, because they are often used to describe processes that grow or decline rapidly on a non-linear scale, such as population growth, radioactive decay, or compound interest.

Here are some examples of how exponential functions can be relevant in business economics:

**Compound interest:** If you invest money in a bank or another form of investment, and your investment grows at a constant rate, the value of your investment will grow exponentially. The formula for future value under compound interest is $FV = P(1+\frac{r}{n})^{nt}$, which is a form of exponential function.

**Economic growth:** When economists talk about long-term growth rates, they often use exponential models. If an economy grows at a constant rate, its size over time will follow an exponential curve.

**Depreciation:** Some types of depreciation, such as declining-balance depreciation, can be modeled with an exponential function, where the asset's value decreases exponentially over time.

It is therefore important to understand what an exponential function is and how to work with it. Being able to model and understand exponential growth or decline can be crucial for many aspects of business economics.

### The logarithm

The logarithm is the inverse function of the exponential function,
i.e. if $y = b^x$, then $x = \log_{b}(y)$

Logarithms are a fundamental concept in mathematics and are widely used in both science and engineering. Logarithms help simplify calculations and make it easier to handle very large or very small numbers.

A logarithm indicates the exponent to which a specific base must be raised to produce a given number. For example, in the expression $10^2 = 100$, 2 is the logarithm of 100 in base 10. We can write this as $\log_{10}(100) = 2$.

**The relationship between logarithms and exponents:** As we have seen, a logarithm is essentially the inverse operation of an exponent. This is useful because it makes it easier to carry out calculations that would be difficult with exponents alone.

**Handling large numbers:** Logarithms can make very large numbers more manageable. For example, instead of working with a value like 1,000,000,000, we can work with its logarithm in base 10, which is 9 - this can be written in Python as $1e9$.

**Multiplication and division become addition and subtraction:** One of the most powerful properties of logarithms is that they can turn multiplication and division into addition and subtraction.

Overall, logarithms are a powerful tool that helps us simplify and handle complex mathematical problems.

In the example above, we define x as a list of numbers from -2 to 2, and we use these values as input to the exponential function $e^x$. The exponential function can take any real number as input, including negative numbers. For example, $e^{-1}$ is equal to about 0.3678.

The result of the exponential function (i.e. $e^x$) is always a positive number, because $e$ (the base of the natural logarithm, about 2.71828) raised to any power can never become 0 or negative. So even though we start with negative values in our x-array, y_exp ($e^x$) is always a positive value.

When we then take the logarithm of y_exp (i.e. np.log(y_exp)), we only use positive values as input to the logarithm, because y_exp contains only positive values. So we get no errors or invalid results, even though the logarithm is undefined for 0 or negative numbers.

In short: even though our original x-values include negative numbers, we only use positive values as input to the logarithm, because we first pass the x-values through the exponential function.

### Example with compound interest:

The formula for compound interest: $FV = P(1+\frac{r}{n})^{nt}$

$FV$ = Future value

$P$ = Principal

$r$ = Interest rate

$n$ = Number of times interest compounds per year

$t$ = Time, measured in years



```python
# Define variables
P = 1000  # Principal (the original loan or investment amount)
r = 0.05  # Annual interest rate (as a decimal, so 5% becomes 0.05)
n = 4     # Number of times interest compounds per year (we assume quarterly compounding here)
t = 20    # Number of years the investment runs
```

```python
# List to store the results
FV_list = []

# Calculate the future value for each year and store the result
for i in range(t+1):
    FV = P * (1 + r/n)**(n*i)
    FV_list.append(FV)
```

`range(n)` gives the numbers $0, 1, ..., n-1$ - that is, $n$ numbers, starting from 0 and stopping BEFORE $n$ (a programming quirk that shows up everywhere in Python).

In Python, indexing of lists and arrays starts from 0. So when we use **range(t+1)** in our for-loop, we ensure that the loop runs from $i = 0$ to $i = t$, which includes $t$ itself. If we used **range(t)**, the loop would only run from $i = 0$ to $i = t-1$.

Here's how the loop works with respect to indexing and calculating $FV$ (Future Value):

* When $i = 0$: $FV = P \times \left(1 + \frac{r}{n}\right)^{0} = P$ (Present value)
* When $i = 1$: $FV$ for the first year.
* When $i = 2$: $FV$ for the second year.
* When $i = t$: $FV$ for the $t$-th year.

If you used **range(t)** instead of **range(t+1)**, you would not get $FV$ for the $t$-th year, since the loop would stop at $t-1$.

So, $t+1$ in **range(t+1)** ensures that all values from year 0 to year $t$ are included in your calculations.

```python
# Print explanation
print(f"Over 20 years we have earned {FV_list[20] - FV_list[0]:.2f}kr in interest. So there is {FV_list[20]:.2f}kr in our account after an original investment of {int(FV_list[0])}kr.")
```

#### Find the original deposit from Future Value

To identify the initial deposit for a given future value, the following formula can be used.

$P = \frac{FV}{\left(1 + \frac{r}{n}\right)^{nt}}$

```python
# Calculate P directly
P = FV / (1 + r/n)**(n*t)

print(f"The original investment was {P:.2f}.")
```

#### Find the amount of time

If t (the number of years) is unknown, we can use the equation $FV = P * (1 + r/n)^{nt}$ to find t.

We isolate t by rewriting this equation as:

$t = (log(FV/P) / log(1 + r/n)) / n$

If we use the original values for $P$, $r$, and $n$, and assume we know the future value ($FV$), we can find $t$ with the following Python code:

```python
# Define known variables
P = 1000  # Principal
r = 0.05  # Annual interest rate (decimal)
n = 4     # Number of times interest compounds per year
FV = 2701.4849407533275  # Future value

# Solve for t using logarithm rules
t = (np.log(FV/P) / np.log(1 + r/n)) / n

print(f"The investment has run for {t:.2f} years.")
```

#### Identify the interest rate

$r = n \left( \left( \frac{FV}{P} \right)^{\frac{1}{nt}} - 1 \right)$

The formula above is designed to calculate the interest rate $r$ based on a known future value ($FV$), the present value of the investment ($P$), the number of interest payments per period ($n$), and the number of periods ($t$). The formula is a rearrangement of the classic future-value formula in the context of compound interest.

Here's what happens:

* $FV/P$ calculates the growth multiplier for the investment.
* $(FV/P)^{1/(nt)}$ calculates the root of this multiplier, adjusted for the number of interest payments and periods.
*  $((FV/P)^{1/(nt)} - 1)$ subtracts 1 from this root to isolate the interest-rate component.
* $n \cdot ((FV/P)^{1/(nt)} - 1)$ finally multiplies by the number of interest payments per period to obtain the actual annual interest rate $r$.

```python
# Assume we know the following values:
P = 1000  # Principal
FV = 2701.4849407533275  # Future value
n = 4  # Number of compounding periods per year
t = 20  # Number of years
```

```python
# Calculate r using the formula:
r = n * ((FV/P)**(1/(n*t)) - 1)
```

```python
# interest rate as a decimal
print("The interest rate is:", r)
```

```python
# interest rate as a percentage
print(f"The interest rate is: {r*100:.2f}%")
```

#### The logarithm - doubling an investment

This code uses the natural logarithm to calculate how long it takes to double an investment at a given annual interest rate $r$. The code uses the precise formula - the so-called "Rule of 72" is a rule of thumb that approximates the result as $72$ divided by the interest rate in percent.

$t = \frac{\ln(2)}{\ln(1 + r)}$

Here's what happens:

* $t$ = np.log(2) / np.log(1 + r): This uses the formula to find $t$, the number of years it takes to double an investment. The formula uses natural logarithms and is derived from the general formula for compound interest. It calculates $t$ by taking the natural logarithm of $2$ (because we want to double the investment) and dividing it by the logarithm of ($1+r$).

"Rule of 72" is a simplification, or rule of thumb, derived from the more precise mathematical formula for compound interest. In the full compound-interest formula, we use exponents to calculate how much an investment will grow over time.

To find the time $t$ it takes for an investment to double, we set $FV=2×P$ in the future-value formula and solve for $t$. This gives us an equation with an exponential variable, which we can solve by taking the logarithm of both sides. In this context we use the natural logarithm for its mathematical simplicity, and because it underlies many growth processes in nature and economics.

So the natural logarithm is used in the precise calculation because it makes it easier to isolate $t$ and find a solution. The "Rule of 72" is a simplification of this precise mathematical model, using the number 72 instead of requiring you to calculate logarithms. It's an easier, but less precise, way to estimate the doubling time of an investment.

```python
# annual interest rate
r = 0.05

# Calculate t using logarithms
t = np.log(2) / np.log(1 + r)

print(f"It takes {t:.2f} years to double your investment.")

# Compare with the Rule of 72 approximation
print(f"The Rule of 72 estimate gives: {72/(r*100):.2f} years.")
```

# Derivatives
In mathematics, "derivative" and "the derived function" mean the same thing. Both terms refer to the operation that measures how much a function changes when its input values change. This concept is fundamental in calculus and is often used to determine the slope of a function at a specific point, or to predict how a variable will change given small changes in another variable.

More formally, the derivative of a function at a given point is defined as the limit of the function's slope between two points, as those two points get infinitely close to each other. If this limit exists, the function is said to be differentiable at that point, and the value of the limit is the function's derivative at that point.

For example, if we have a function $f(x) = x^2$, then its derivative is $f'(x) = 2x$, which means that for every small change in x, $f(x)$ changes by approximately $2x$ times the size of that change.

diff is a function in the SymPy library used to calculate the derivative of a function. The derivative measures how a function changes when its input values change. In practice, it is the slope of a tangent line at a specific point on a curve.

diff can be used for several things, including:

* diff(f, x): Returns the first derivative of f with respect to x.
* diff(f, x, n): Returns the n-th derivative of f with respect to x.
* diff(f, x, y): Returns the first derivative of f with respect to x and then y (i.e. the mixed derivative).

For example, if we have a function $f = x^2 + 2x + 1$, we can calculate its first derivative by writing diff(f, x), which gives $2*x + 2$.

SymPy's diff function can handle many types of functions, including polynomial, trigonometric, exponential, logarithmic, and composite functions, and it can be applied using many different differentiation rules, including the product rule, the quotient rule, and the chain rule

### The chain rule
The chain rule is a fundamental rule in differential calculus, used to find the derivative of composite functions. A composite function is a function "built up" from two or more functions. For example, if we have a function $ f(x) $ and another function $ g(x) $, then $ f(g(x)) $ is a composite function.

Put simply, the chain rule helps us break a composite function down into its "building blocks" and differentiate each piece separately. We then multiply these derivatives together to get the overall derivative.

Mathematically, the chain rule is defined as follows:
$\frac{d}{dx}[f(g(x))] = f'(g(x)) \times g'(x)$

Here $ f'(g(x)) $ represents the derivative of $ f $ evaluated at $ g(x) $, and $ g'(x) $ is the derivative of $ g $.


```python
# We import symbols, diff, exp, and cos from sympy - we do this so we can call the functions directly without having to reference the sympy library
# By running this code, we can therefore use symbols, differentiation, the exponential function, and cosine from the SymPy package.
from sympy import symbols, diff, exp, cos, solve
```

#### Example:

$f(x) = cos(x^2)$

The chain rule is needed here, because $f(x)$ is a composite function ($cos$ of $x^2$). SymPy's .diff() method automatically detects when it should apply the chain rule (as well as other differentiation rules such as the product rule and the quotient rule), depending on the function you're trying to differentiate. This makes it very easy to work with complex or composite functions, since the library handles all the technical details for us. We simply give it the function we want to differentiate, and the variable with respect to which we want to take the derivative, and .diff() does the rest.

```python
# Define x as a symbol
x = symbols('x')

# Define function f
f = cos(x**2)
```

```python
# Calculate the derivative of cos(x^2)
derivative_f = f.diff(x)

# Show the derivative
derivative_f
```

```python
# Find the derivative when x = .5
at_05 = derivative_f.subs(x, .5)
at_05
```

### The product rule

The product rule is a method in differential calculus for finding the derivative of a product of two functions. Suppose you have two functions $f(x)$ and $g(x)$. The product rule states that the derivative of their product is:

\begin{equation}
(fg)'(x) = f'(x)g(x) + f(x)g'(x)
\end{equation}


Here $ f'(x) $ and $ g'(x) $ represent the derivatives of $ f(x) $ and $ g(x) $ with respect to $ x $, and $ fg $ represents the product of $ f $ and $ g $.

In other words, to find the derivative of a product of two functions, you take the first function and multiply it by the derivative of the second function, and add that to the second function multiplied by the derivative of the first function.

The product rule is very useful in many areas of mathematics, physics, and engineering, where you often need to find the derivative of a product of two functions.


#### Example:
$f(x) = exp(x)*cos(x)$

```python
# Define x as a symbol (just to be safe)
x = symbols('x')

# Define function f
f = exp(x)*cos(x)
```

```python
# Calculate the derivative
derivative_f = f.diff(x)
derivative_f
```

```python
# Find the derivative when x = .7
at_07 = derivative_f.subs(x, .7)
at_07
```

### The power rule
The power rule is a method in differential calculus for finding the derivative of a function of the form $ f(x) = x^n $, where $ n $ is a constant. According to the power rule, the derivative of $ f(x) $ is given by:

\begin{equation}
f'(x) = n \times x^{(n-1)}
\end{equation}

This formula says that to find the derivative of $ x^n $, you multiply the exponent $ n $ by $ x $ raised to $ n-1 $. The power rule is very versatile and forms the basis for finding the derivative of a wide range of functions. It is especially useful because it allows you to quickly differentiate a broad class of functions containing powers of $ x $.






#### Example:

$f(x) = x^5$

```python
# We define x again, just to be safe
x = symbols('x')

# Define function f
f = x**5

# Calculate the derivative
derivative_f = f.diff(x)
derivative_f
```

```python
derivative_f.subs(x, 5)
```

### The quotient rule
The quotient rule is a method in differential calculus used to find the derivative of a quotient of two functions, $ f(x) $ and $ g(x) $. According to the quotient rule, the derivative of the quotient $ \frac{f(x)}{g(x)} $ is given by:

\begin{equation}
\left( \frac{f(x)}{g(x)} \right)' = \frac{f'(x)g(x) - f(x)g'(x)}{(g(x))^2}
\end{equation}

This rule is especially useful when working with functions that are divided by one another, where you want to find the derivative of such expressions. Essentially, you take the derivative of the numerator function and multiply it by the denominator function, then subtract the product of the numerator function and the derivative of the denominator function. The whole thing is then divided by the square of the denominator function.






#### Example:

$f(x) = x^2$ and $g(x) = x + 1$

```python
x = symbols('x')
```

```python
# Define our functions f and g
f = x**2
g = x + 1
```

```python
# Apply the quotient rule to find the derivative of h = f / g
h_prime = (g * diff(f, x) - f * diff(g, x)) / g**2
h_prime
```

```python
solve(h_prime)
```

Here we apply the quotient rule to find the derivative of a quotient of two functions $f(x)$ and $g(x)$.

The derivative of $f(x)=x^2$ is $f′(x)=2x$, and the derivative of $g(x)=x+1$ is $g′(x)=1$.

Substituting these into the quotient rule, we get:

$$h' = \frac{(x+1) \cdot 2x - x^2 \cdot 1}{(x+1)^2}$$

solve($h′$) finds the roots of the derivative function $h′$, which gives the values of $x$ where $h$ has local extrema (minimum or maximum).

The roots $[−2,0]$ of the derivative function $h′$ indicate the $x$-values where the function $h(x)$ has local extrema. In practice, this means the function has either a local minimum or maximum at these points, or it could be a "saddle" point.

* x$=−2$ and $x=0$ are the values where the slope of $h(x)$ is 0.
* To determine whether it's a minimum or maximum, you can use the second-derivative test, or simply look at the function on either side of the $x$-values.

So in this context, if we want to find out more about the behavior of $h(x)$ at these points, you can examine the values of $h(x)$ close to $x=−2$ and $x=0$. This will give you an idea of whether the function is increasing or decreasing near these points, and thus whether we're dealing with local minima or maxima.
