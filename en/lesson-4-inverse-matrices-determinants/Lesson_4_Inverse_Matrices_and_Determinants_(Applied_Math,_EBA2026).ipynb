In a square matrix equation, as we saw in lesson 2, we try to solve
for the vector $\vec{x}$.

$A\vec{x} = \vec{b}$

Here the matrix $A$ is multiplied by the vector $\vec{x}$, so we need a way to remove $A$.

Let's look at the identity matrix: it consists of 1s along the diagonal and 0s everywhere else. Here we see a 2x2 identity matrix, which we will write as $I_2$:

$I_2 =
  \begin{bmatrix}
    1 & 0\\
    0 & 1 \\
  \end{bmatrix}$

When we multiply $I_2$ by a vector containing 2 elements, the resulting vector matches the original vector exactly:

$I_2\vec{x} = \vec{x}$

This is because each element in the vector is multiplied exactly once by the
diagonal $1$ value in the identity matrix:

Example with numbers:
$$
  \begin{bmatrix}
    1 & 0 \\
    0 & 1 \\
  \end{bmatrix}
%
  \begin{bmatrix}
    3 \\
    2 \\
  \end{bmatrix}
 %
=
  \begin{bmatrix}
    1*3 + 0*2 \\
    0*3 + 1*2 \\
  \end{bmatrix}
%
=
  \begin{bmatrix}
    3 \\
    2 \\
  \end{bmatrix}
$$

General case:
$$
  \begin{bmatrix}
    1 & 0 \\
    0 & 1 \\
  \end{bmatrix}
%
  \begin{bmatrix}
    x_1 \\
    x_2 \\
  \end{bmatrix}
 %
=
  \begin{bmatrix}
    1x_1 + 0x_2 \\
    0x_1 + 1x_2 \\
  \end{bmatrix}
%
=
  \begin{bmatrix}
    x_1 \\
    x_2 \\
  \end{bmatrix}
$$

If we can transform matrix $A$ and convert it into the identity matrix, only the solution vector $\vec{x}$ will remain.

Let's practice working with the identity matrix before we try to solve the transformation of $A$ into $I$.

We can create any identity matrix using the function `numpy.identity()`. This function has only 1 parameter, `n`, which specifies the $n * n$ identity matrix we need.

## The Inverse Matrix

Now we know a bit more about the identity matrix and can consider how we cancel out
the coefficient matrix $A$. Put another way: we will try to transform $A$ into the identity matrix $I$. This is done by multiplying the matrix by its **inverse** matrix.

An inverse matrix corresponds to the idea of a multiplicative inverse. Suppose, for example, we want to solve for $x$ in the equation $5x = 10$. To do this, we
need to multiply both sides by the multiplicative inverse of $5$, which is $5^{-1}$ (or $1/5$):

$5^{-1}*5x = 5^{-1} * 10$

The inverse of $5$ turns it into $1$ and leaves us with the solution: $x = 2$. To solve for the vector $\vec{x}$ in the matrix equation, we need to multiply both sides by the inverse of $A$:

$A^{-1}A\vec{x}=A^{-1}\vec{b}$

This can be simplified to $I\vec{x}=A^{-1}\vec{b}$, which gives us the formula for calculating the solution vector:

$\vec{x} = A^{-1}\vec{b}$

Although we use the *inverse matrix* to cancel out specific things in essentially the same way as when using a multiplicative inverse, the calculation is completely different. Let's understand the calculation of the inverse of a 2 * 2 matrix.

If $ A =
  \begin{bmatrix}
    a & b\\
    c & d \\
  \end{bmatrix}$
then $A^{-1} = \frac{1}{ad-bc}\begin{bmatrix}
    d & -b\\
    -c & a \\
  \end{bmatrix}$


The expression $ad - bc$ is known as the **determinant** and is often written
as $det(A) = ad - bc$ or as $|A| = ad - bc$. Because we divide by the determinant when calculating the inverse matrix, a 2 x 2 matrix is only invertible if the determinant is not equal to 0.

In this step and the next step we will focus on
finding the inverse matrix when $A$ is a 2 x 2 matrix. Later we will go through how to calculate the inverse matrix for a higher-dimensional matrix (3 x 3 and above).

```python
# What's happening "under the hood" in NumPy
# Manually implemented function to compute the inverse matrix (of a 2x2 matrix)
def matrix_inverse_two(M):
  det = M[0,0]*M[1,1] - M[0,1]*M[1,0]
  if det == 0:
    raise ValueError('Determinant = 0 - this cannot be done')
  else:
    return 1/det * np.array([[M[1,1],-M[0,1]],
                             [-M[1,0],M[0,0]]])
```

```python
# Manual computation of the inverse matrix
inverse_a = matrix_inverse_two(matrix_a)
```

### The Determinant and Inverse Matrix of Higher-Dimensional Matrices

Before we look at how to calculate the inverse matrix for higher-dimensional matrices, we first need to dig deeper into the determinant and introduce some more terminology. So far we have mostly worked with matrices that contain the same number of rows and columns. These matrices are known as square matrices, and we can only calculate the determinant and the inverse matrix for square matrices. In addition, we can only calculate the inverse matrix of square matrices when the determinant is **not** equal to 0.

To find the determinant of a higher-dimensional square matrix, we need
the more general form of the determinant. It looks like this:

$ det(
  \begin{bmatrix}
    a & b & c\\
    d & e & f \\
    g & h & i
  \end{bmatrix}) = a* det
  (\begin{bmatrix}
    e & f\\
    h & i \\
  \end{bmatrix})
  - b*det(
  \begin{bmatrix}
    d & f\\
    g & i \\
  \end{bmatrix}) + c* det
  (\begin{bmatrix}
    d & e\\
    g & h \\
  \end{bmatrix})

The determinant of a higher-dimensional system involves breaking down the full
matrix into smaller matrices. First we choose a row or column (often the first row is chosen). For the first value in the chosen row we "hide" the other values in that row (the 2nd and 3rd values in the row) and in the corresponding column (the 2nd and 3rd values in the column), use the remaining elements as the smaller matrix, and multiply the scalar value by the determinant of the smaller matrix. We repeat this for the remaining values in the first row.

Here is a concrete example:

$ det(
  \begin{bmatrix}
    1 & 2 & 3\\
    4 & 5 & 6 \\
    7 & 8 & 9
  \end{bmatrix}) = 1* det
  (\begin{bmatrix}
    5 & 6\\
    8 & 9 \\
  \end{bmatrix})
  - 2*det(
  \begin{bmatrix}
    4 & 6\\
    7 & 9 \\
  \end{bmatrix}) + 3* det
  (\begin{bmatrix}
    4 & 5\\
    7 & 8 \\
  \end{bmatrix})

To calculate the determinant in NumPy we use the function **numpy.linalg.det()**. More about this function can be found in the [documentation](https://numpy.org/doc/stable/reference/generated/numpy.linalg.det.html), where you can learn more about how to use this function.

To calculate an inverse matrix for a 3 * 3 or larger matrix, we also need to work with the more general form of the matrix's inverse equation. Similar to the determinant for higher-dimensional matrices, the inverse matrix works by generating smaller matrices that depend on the position in the matrix. Here is a diagram describing the inverse matrix for a 3 * 3 matrix:

$
\begin{bmatrix}
    a & b & c\\
    d & e & f \\
    g & h & i
\end{bmatrix} ^{-1} =
\frac{1}{det(\begin{bmatrix}
    a & b & c\\
    d & e & f \\
    g & h & i
\end{bmatrix})}
\begin{bmatrix}
    det(\begin{bmatrix}
    a & - & -\\
    - & e & f \\
    - & h & i \end{bmatrix})  
  & ...
  & det(\begin{bmatrix}
    - & - & c\\
    d & e & - \\
    g & h & - \end{bmatrix}) \\
... & ... & ... \\
det(\begin{bmatrix}
    - & b & c\\
    - & e & f \\
    g & - & - \end{bmatrix})  
  & ...
  & det(\begin{bmatrix}
    a & b & -\\
    d & e & - \\
    - & - & i \end{bmatrix}) \\
\end{bmatrix}
$

Although it's useful to know how to calculate the inverse this way for higher-dimensional matrices, the amount of careful arithmetic you have to do by hand is substantial. Fortunately, the function **numpy.linalg.inv()** can work with any n-dimensional square matrix.

A matrix is a rectangular table of numbers, symbols, or expressions, arranged in rows and columns. Matrices are used to represent linear transformations, systems of linear equations, and many other concepts within linear algebra and mathematics in general.

The inverse of a matrix A, usually denoted A^-1, is a special form of matrix. If you multiply a matrix by its inverse, you get the identity matrix (which is mathematics' equivalent of the number 1 within matrix multiplication).

For example, if we have a matrix A:

$A = [[a, b],
[c, d]]$

its inverse is:

$A^-1 = 1/(ad-bc) * [[d, -b],
[-c, a]]$

Where "1/(ad-bc)" is the reciprocal (or "inverse") of the determinant of matrix A.

**Note** that not all matrices have an inverse. Only square matrices (i.e. matrices with an equal number of rows and columns) can have an inverse, and even then only the matrices whose determinant is not zero have an inverse. These matrices are called invertible or non-singular.

The inverse matrix plays a central role in many applications of linear algebra, including solving systems of linear equations, calculating matrix powers, and many other applications.

## More on the Determinant

A determinant is a special value that can be calculated from the elements of a square matrix. The determinant has a range of properties and many applications within mathematics. The most common place to encounter determinants is in systems of linear equations, since it helps determine whether the system has a unique solution.

Let's look at a 2x2 matrix:

A = [[a, b],
[c, d]]

The determinant of this matrix, often written as |A| or det(A), is calculated as:

|A| = ad - bc

The determinant has a number of interesting properties. For example, if the determinant of a matrix is 0, we say that the matrix is "singular" or "degenerate", meaning it does not have an inverse. This ties into the idea we discussed earlier - a matrix only has an inverse if the determinant is not 0.

For larger matrices, calculating the determinant is a bit more complex, but it still follows patterns based on the elements of the matrix.

In general, the determinant is important because it tells us a lot about the matrix, including whether we can find its inverse. This is critical for many applications of linear algebra, including solving systems of linear equations.

## Input-Output Models (the Leontief Model)

The expression $(I - M)^{-1}$ refers to the inverse matrix of $(I−M)$, where $I$ is the identity matrix and $M$ is the input matrix for the economy. This is a central part of the so-called Leontief Input-Output Model, a method for describing how different sectors of an economy interact with each other.

In practice, $(I - M)^{-1}$ is used to calculate the total output $(X)$ from each sector needed to meet a given final demand $(D)$. Mathematically, this is given by:

$X=(I−M)^{−1} ×D$

Here $D$ represents the final demand from each sector, and $X$ is the total output from each sector needed to meet this demand.

Why is this important?

* It accounts for the interdependence between different sectors. For example, if sector A requires input from sector B for its production, and sector B in turn requires input from sector A, both of these relationships need to be included in the analysis.

* It provides a way to calculate how changes in demand in one sector will affect output and demand in other sectors.

* It is especially useful for policymakers, economists, and business leaders in understanding how changes in one sector can have a cascading effect on the whole economy.

#### Example with the Leontief Model

An economy is based on three sectors: trade (H), transport (T), and health (S). Producing one krone's worth of trade requires an input of 0.25 DKK from the trade sector and 0.15 DKK from the transport sector. Producing one krone's worth of transport requires an input of 0.05 DKK from the transport sector and 0.35 DKK from the health sector. Producing one krone's worth of health requires an input of 0.20 DKK from the trade sector, 0.05 DKK from the transport sector, and 0.25 DKK from the health sector.

* Find the input matrix $M$ for this economy
* Calculate $(I - M)^{-1}$ using Python
* Find the output from each sector needed to meet a final demand of 15 billion DKK for trade, 5 billion DKK for transport, and 25 billion DKK for health.

```python
# Define the input matrix, M
M = np.array([
    [0.25, 0.15, 0],
    [0, 0.05, 0.35],
    [0.20, 0.05, 0.25]
])
```

```python
# Calculate (I - M)^{-1}
I = np.identity(3)  # 3x3 Identity matrix
IM_inv = np.linalg.inv(I - M)
```

```python
# Calculate output for each sector
D = np.array([15, 5, 25])  # vector for final demand
X = IM_inv @ D
```

```python
print("Input Matrix M:")
print(M)
print("\n (I - M)^{-1}:")
print(IM_inv)
print("\n Output from each sector, needed to meet demand:")
print(X)
```

## Example with a Linear System

$A\vec{x} = \vec{b}$

$$
\text{The system of equations is given by:}
\begin{aligned}
\begin{smallmatrix} \textrm{A}\\\begin{bmatrix}
2 & 1 & 4 & 1 \\
3 & 4 & -1 & 2 \\
1 & -1 & 3 & 2 \\
2 & 7 & 1 & 5 \\
\end{bmatrix}\end{smallmatrix}
*
\begin{smallmatrix} \textrm{x}\\\begin{bmatrix}
x_1 \\
x_2 \\
x_3 \\
x_4 \\
\end{bmatrix}\end{smallmatrix}
&=
\begin{smallmatrix} \textrm{B}\\\begin{bmatrix}
15 \\
10 \\
13 \\
23 \\
\end{bmatrix}\end{smallmatrix}
\end{aligned}
$$

$$
\text{It can also be expressed as individual equations: }
\begin{aligned}
2x_1 + 1x_2 + 4x_3 + 1x_4 &= 15, \\
3x_1 + 4x_2 - 1x_3 + 2x_4 &= 10, \\
1x_1 - 1x_2 + 3x_3 + 2x_4 &= 13, \\
2x_1 + 7x_2 + 1x_3 + 5x_4 &= 23.
\end{aligned}
$$

```python
# Define a 4x4 matrix A and a column vector B
A = np.array([[2, 1, 4, 1],
              [3, 4, -1, 2],
              [1, -1, 3, 2],
              [2, 7, 1, 5]])

B = np.array([15, 10, 13, 23])
```

```python
# Calculate the determinant of A
det_A = np.linalg.det(A)
```

```python
# Check whether the determinant is different from 0 (if it is 0, the matrix has no inverse)
if det_A == 0:
    print("The matrix has no inverse.")
else:
    # Calculate the inverse matrix of A
    A_inv = np.linalg.inv(A)

    # Solve for X by multiplying A_inv by B
    X = A_inv @ B

    print("The solution set X is:", X)
```
