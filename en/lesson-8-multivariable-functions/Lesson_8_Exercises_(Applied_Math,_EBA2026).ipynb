#### A brief introduction to the Lagrange method

The Lagrange method is used when we want to minimize (or maximize) a function *while* a constraint must be satisfied - here: minimize the cost $C(x, y)$, while production satisfies $4x + 2y = 100$.

The trick is to combine everything into one new function, the **Lagrange function**:

$$L(x, y, \lambda) = C(x, y) - \lambda \cdot (4x + 2y - 100)$$

The new variable $\lambda$ ("lambda") keeps track of the constraint: the parenthesis is exactly 0 precisely when the constraint is satisfied. From here we do what we usually do with critical points - we find the partial derivatives of $L$ (now with respect to all three variables) and set them equal to 0. The solution is the point that gives the lowest possible cost without breaking the constraint.

In economics, $\lambda$ even has its own interpretation: it shows roughly how much the cost changes if the constraint is relaxed by one unit (e.g. if the production requirement is raised from 100 to 101). You will encounter that interpretation again later in your studies.
