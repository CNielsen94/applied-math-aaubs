As we know, a matrix consists of column vectors. Because of this, we can use the same concepts we learned about matrix-vector multiplication to multiply matrices with each other. In matrix-vector multiplication, we performed a dot product between each row of the first matrix and the column vector. In matrix multiplication, we extend this to perform a dot product between each row of the first matrix and each column of the second matrix.

$$
AB =
  \begin{smallmatrix} \textrm{A}\\\begin{bmatrix}
    0.7 & 3 \\
    1.7 & 2 \\
  \end{bmatrix}\end{smallmatrix}
·
 \begin{smallmatrix} \textrm{B}\\\begin{bmatrix}
    1 & 2\\
    2 & 3\\
  \end{bmatrix}\end{smallmatrix}
=
\begin{smallmatrix} \textrm{A * B}\\\begin{bmatrix}
    (0.7*1+3*2) & (0.7*2+3*3) \\
    (1.7*1+2*2) & (1.7*2+2*3) \\
  \end{bmatrix}\end{smallmatrix}
=
  \begin{smallmatrix} \textrm{AB}\\\begin{bmatrix}
    6.7 & 10.4 \\
    5.7 & 9.4 \\
  \end{bmatrix}\end{smallmatrix}
$$

Just like with matrix-vector multiplication, the columns of the first matrix must match the number of rows in the second matrix.

```python
A = np.array([[0.7, 3],
              [1.7 , 2]])

B = np.array([[1, 2],
              [2 , 3]])

A @ B
```

NOTE: Order matters!

$$
BA =
  \begin{smallmatrix} \textrm{B}\\\begin{bmatrix}
    1 & 2 \\
    2 & 3 \\
  \end{bmatrix}\end{smallmatrix}
·
 \begin{smallmatrix} \textrm{A}\\\begin{bmatrix}
    0.7 & 3\\
    1.7 & 2\\
  \end{bmatrix}\end{smallmatrix}
=
\begin{smallmatrix} \textrm{B * A}\\\begin{bmatrix}
    (1*0.7+2*1.7) & (1*3+2*2) \\
    (2*0.7+3*1.7) & (2*3+3*2) \\
  \end{bmatrix}\end{smallmatrix}
=
  \begin{smallmatrix} \textrm{BA}\\\begin{bmatrix}
    4.1 & 7 \\

  \end{bmatrix}\end{smallmatrix}
$$

The matrix transpose swaps the rows and columns of a matrix. You can think of the transpose as a rotation. In data analysis, statistics, and finance, data tables of different dimensions are often used. Because of the requirements for matrix multiplication, we sometimes need to transpose a matrix so that we can multiply matrices with each other that don't have the same number of rows and columns.

A matrix transpose will typically look like this:

Original matrix:
$$
  \begin{bmatrix}
    1 & 2\\
    3 & 4 \\
    5 & 6 \\
  \end{bmatrix}
$$

After transposing:

$$
  \begin{bmatrix}
    1 & 3 & 5\\
    2 & 4 & 6\\
  \end{bmatrix}
$$

Mathematically, we use the notation $A^T$ to denote the transpose operation.

$A^T + B^T = C$

The transpose has some interesting rules that are quite intuitive. For example, when we take the transpose of the sum of two matrices, we can distribute the transpose to each matrix:

$(A + B)^T = A^T + B^T$

It's a bit more counterintuitive, however, when taking the transpose of the product of two matrices:

$(AB)^T = B^T A^T$

Let's see this in NumPy: we can either use `np.transpose()` or simply write `.T` after a matrix.

##### **Key concepts**
*   Many operations that can be performed on vectors can also be performed on matrices. We can perform the following operations with matrices:

##### NumPy's random function

**Random seed**

You can use random numbers from NumPy and still get the same results every time, if you use

`np.random.seed(42)`

```python
# Bonus: random numbers in NumPy - with np.random.seed() "random" numbers become reproducible
np.random.seed(42)
np.random.rand(2,2)
```

```python
# Same seed gives EXACTLY the same numbers - try comparing with the cell above
np.random.seed(42)
np.random.rand(2,2)
```

**Random.rand**

`np.random.` is used for random sampling and is often applied in mathematics, statistics, finance, and various scientific applications.

For example, `np.random.rand()` is used to generate random values based on different parameters. The following are examples of how `np.random` can be used in different situations, but more information can be found [here](https://numpy.org/doc/stable/reference/random/index.html).

```python
# Generate a single random value
random_value = np.random.rand()
print(f"Single random value: {random_value}")

print("\n")

# Generate an array with 5 random values
random_array = np.random.rand(5)
print(f"Array with 5 random values: {random_array}")

print("\n")

# Generate a 2x2 matrix with random values
random_matrix = np.random.rand(2, 2)
print(f"2x2 matrix with random values: \n{random_matrix}")
```

```python
np.random.seed(42)

# Generate a random integer between 1 and 10
random_integer = np.random.randint(1, 11)
print(f"Random integer between 1 and 10: {random_integer}")

print("\n")

# Generate an array with 5 random integers between 1 and 10
random_integer_array = np.random.randint(1, 11, 5)
print(f"Array with 5 random integers between 1 and 10: {random_integer_array}")

print("\n")

# Generate a 3x3 matrix with random integers between 1 and 10
random_integer_matrix = np.random.randint(1, 11, (3, 3))
print(f"3x3 matrix with random integers between 1 and 10: \n{random_integer_matrix}")
```

```python
# Randomly shuffle an array
arr = np.array([1, 2, 3, 4, 5])

np.random.shuffle(arr)


print(f"Shuffled array: {arr}")
print("\n")

# Generate a random permutation
perm = np.random.permutation(5)
print(f"Random permutation: {perm}")
print("\n")

random_permutation = np.random.permutation(arr)
print(f"Random permutation of array: {random_permutation}")
```

`np.random.permutation` does not have built-in functionality to specify a starting point or number of elements directly as parameters. The function either takes an integer `x` and returns a random permutation of `np.arange(x)` (which starts at 0 and goes up to `x-1`), or it takes an existing array and returns a random permutation of that array.

```python
#np.random.seed(42)

# Create a list of options
options = ['apple', 'banana', 'lemon', 'pineapple', 'pear', 'orange', 'mandarin']

# Choose a random element from the list
random_choice = np.random.choice(options)

print(f"Randomly chosen fruit: {random_choice}")
```

```python
# Choose from options with different probabilities
random_choice_with_probability = np.random.choice(options, p=[0.1, 0.2, 0.3, 0.1, 0.1, 0.1, 0.1])
print(f"Fruit chosen with specified probability: {random_choice_with_probability}")
```

Remember that the parameter `p` refers to probability. The values in `p` must both match the number of observations in the given list and sum to 1. In general, it's a good idea to understand the hyperparameters and functionality of the different packages you use in Python. It's often possible to achieve significant optimization by using parameters correctly and at the right time.
