## What is a program?

Imagine you have to explain to a friend how to make a cup of coffee. You would give a series of very precise instructions:

## Module 1: Named Boxes (Variables and Data Types)

So far, we've only asked the computer to write text that we gave it directly in the `print()` command. But what if we want to store information and use it again later? Imagine you're writing a program that needs to remember a user's name, age, or score in a game. It would be impractical to have to type the information in again and again.

This is where variables come in.

### The Analogy: A Named Box

The easiest way to think about a variable is as **a named box**. It's a container where you can store one piece of information.

*   **The box has a name** (the variable name), so you can easily find it again.
*   **You can put something in the box** (you assign it a value).
*   **You can look inside the box** to see what's in it (you use the variable in your code).
*   **You can replace the contents** with something new (you can change the variable's value).

In Python, we create a variable (a box) by giving it a name and using the equals sign (`=`) to put a value into it.

`box_name = value`

#### Good habits: naming variables
Giving your "boxes" good names is one of the most important habits to learn. It makes your code much easier to read, for yourself and for others. Here are a few simple rules:
*   **Use descriptive names:** `customer_age` is much better than `x`.
*   **Use `snake_case`:** In Python, the convention is to write variable names in lowercase and separate words with an underscore (`_`). E.g. `my_first_variable`.
*   **Don't start with a number:** `1_car` is an invalid name, but `car_1` is fine.
*   **Avoid special characters:** Stick to letters, numbers, and underscores.

Let's look at the different types of "things" we can put in our boxes. These are called **data types**. We'll start with the one you've already met: text.

### Strings (Text)

In programming, a piece of text is called a **string**. It can be a single letter, a word, a sentence, or an entire book. The important thing is that we always have to tell Python that something is a string, by wrapping it in quotation marks (`" "`) or apostrophes (`' '`).

#### Create and use a string variable

Here I create a variable (a box) named `greeting`. I put the text "Hello, Denmark!" into it. Then I use `print()` to look inside the box and write the contents out on the screen.

````
# 1. Create a variable named 'greeting' and assign it a text value.
greeting = "Hello, Denmark!"

# 2. Use print() to show the value stored in the variable 'greeting'.
print(greeting)
````

#### Interpretation

As you can see, the output is `Hello, Denmark!`. Notice that we didn't write `print("Hello, Denmark!")`. Instead we wrote `print(greeting)`. Because `greeting` isn't in quotation marks, Python understands that it's not the literal text "greeting" we want to write out. Instead, it understands that it should find the box named `greeting` and show us what's inside. This is incredibly useful, because now we can change the contents of `greeting` in one place, and everywhere we use the variable will automatically be updated.

### A smarter way to print: f-strings

Often we want to print a sentence that mixes plain text with the values from our variables. Using `+` can quickly get messy. Python has a fantastic feature for this, called an **f-string**.

Here's how it works:

1.  Start your text string with the letter `f` right before the first quotation mark.
2.  When you want to insert the value of a variable, you write the variable's name inside curly braces `{}`.

#### Use an f-string

See how much nicer and more readable it is to build a sentence with an f-string.

````
name = "Anders"
age = 42

# Using an f-string to combine text and variables
print(f"Hi, my name is {name}, and I am {age} years old.")
````

#### What happened?

The output is a neat, coherent sentence. Python automatically replaced `{name}` with the contents of the `name` variable ("Anders") and `{age}` with the contents of the `age` variable (42). F-strings are the modern and preferred way to format text in Python.

### Numbers: Integers (Whole Numbers) & Floats (Decimal Numbers)

Of course we can store more than just text in our boxes. One of the most common things to work with in programming is numbers. Python has two primary number types:

1.  **Integers** (whole numbers): These are whole numbers without decimals, e.g. `10`, `-5`, `0`, `1234`.
2.  **Floats** (decimal numbers): These are numbers with decimals. **Important:** In Python (and most other programming languages) we use a **period `.`** as the decimal separator, not a comma `,`. E.g. `3.14`, `-0.5`, `99.95`.

When you assign a numeric value to a variable, you should **not** use quotation marks.

*   `my_age = 29` (This is the number 29, an integer)
*   `my_city = "29"` (This is the text "29", a string)

With numbers, we can of course perform mathematical calculations. The most common operators are:
* `+` (plus)
* `-` (minus)
* `*` (times)
* `/` (divide)

#### Calculations with number variables

Let's say we're going to the supermarket. I'll create variables for the quantity of an item and the price per unit. Then I calculate the total price.

````
# Creates an integer variable
num_milk_cartons = 3

# Creates a float variable (remember the period!)
price_per_carton = 12.50

# Calculates the total price by multiplying the two variables
total_price = num_milk_cartons * price_per_carton

# Uses an f-string to print the result nicely
print(f"That'll be {total_price} kr. in total, please.")
````

#### Interpretation

The program calculated `3 * 12.50` and stored the result, `37.5`, in the variable `total_price`. That value was then inserted into our f-string. The result of a calculation between an integer and a float will always be a float.

### An important skill: changing type

Sometimes we have data in one type but need to use it as another. Imagine you have the number `25` stored as a string: `age_as_text = "25"`. If you try to add 1 to it, you'll get an error, because you can't add a number to text.

Python gives us simple functions to convert between types:
* `int()`: Tries to turn something into an integer.
* `float()`: Tries to turn something into a float.
* `str()`: Turns something into a string.

#### Conversion in practice

````
# Here we have a number stored as text
postal_code_text = "2100"
print(f"The variable 'postal_code_text' contains the text: {postal_code_text}")

# We can convert it to a real number, so we can do math with it
postal_code_number = int(postal_code_text)
print(f"After converting with int() it's now the number: {postal_code_number}")

# Now we can perform a calculation
print(f"Calculation: {postal_code_number} + 100 = {postal_code_number + 100}")
````

#### What happened?
We used `int()` to convert a string containing a whole number into a real integer we can do math with. Conversely, we can use `str()` to convert a number into a string, e.g. if we want to combine it with other text using `+`. This is a critical skill, especially when working with user input, which often comes in as text.

### Bridge to mathematics: A night out on Jomfru Ane Gade

Now that you can work with numbers and variables, we can build a bridge to something you know from mathematics: linear functions, e.g. `y = a*x + b`.

Imagine you're planning a night out in Aalborg on Jomfru Ane Gade - Aalborg's famous bar street - and have to choose between two places:
1.  **Bar i Gaden:** Here there's a 50 kr. entry fee, but a beer only costs 30 kr.
2.  **A cosy corner pub:** Here entry is free, but a beer costs 45 kr.

We can model this with two linear functions, where `y` is the total price for the evening, and `x` is the number of beers you buy.
* `y_street = 30*x + 50`
* `y_pub = 45*x + 0`

Let's use Python to calculate what the evening will cost for a given number of beers.

````
# We define a variable for the number of beers
num_beers = 10

# Calculate the price for an evening at Bar i Gaden
entry_fee_street = 50
price_per_beer_street = 30
total_price_street = price_per_beer_street * num_beers + entry_fee_street

# Calculate the price for an evening at the pub
entry_fee_pub = 0
price_per_beer_pub = 45
total_price_pub = price_per_beer_pub * num_beers + entry_fee_pub

print(f"After drinking {num_beers} beers:")
print(f"- Price for an evening at Bar i Gaden: {total_price_street} kr.")
print(f"- Price for an evening at the pub: {total_price_pub} kr.")
````

#### What happened?

We just used our knowledge of variables and simple calculations to model a real-world problem. By changing the value of `num_beers` you can quickly calculate and compare the two venues. Try setting `num_beers` to `2` and then to `4` and see when one place becomes cheaper than the other. This is an example of how programming can be used to analyze and solve mathematical problems.

As some of you may remember from math class, there are ways to visualize these relationships, which is often more intuitive to read than raw numbers. For this, we need to create a series of x-values (different numbers of beers) that we can use to calculate the y-value (the total price).

Here we use `numpy`'s `arange` function. `arange` is literally short for "a range". This means it generates a series of numbers based on the parameters we set.

`np.arange(start, stop, step)`: `start` = the start value, `stop` = the stop value, `step` = how big a "step" we take, i.e. 1 at a time.

For example, `np.arange(0, 11, 1)` generates all the values from 0 up to 10 as integers.

````
import numpy as np
np.arange(0,11,1)
````

````
# Step 1: Import the toolboxes
import matplotlib.pyplot as plt
import numpy as np

# Step 2: Create a range of x-values (whole numbers of beers) from 0 to 10
# np.arange(start, stop, step) stops BEFORE 'stop', so we use 11 to include 10.
num_beers_x = np.arange(0, 11, 1)

# Step 3: Calculate the y-values (total price) for each venue
# y = 30x + 50
price_street_y = 30 * num_beers_x + 50
# y = 45x
price_pub_y = 45 * num_beers_x

# Step 4: Plot both lines. We use 'o-' to show the actual data points.
plt.plot(num_beers_x, price_street_y, 'o-', label="Bar i Gaden")
plt.plot(num_beers_x, price_pub_y, 'o-', label="Pub")

# Step 5: Add titles and labels to make the graph easy to read
plt.title("Price comparison: the Street vs. the Pub (whole beers)")
plt.xlabel("Number of beers bought")
plt.ylabel("Total price (kr.)")

# Step 6: Add a 'legend' to show which line is which
plt.legend()
# Step 7: Add a grid to improve readability
plt.grid(True)

# Step 8: Show the graph
plt.show()
````

### Very basic lists
One last building block you will need from the start: a **list** collects several values in one variable. We go much deeper in Supplement B (loops) and Supplement C (lists, dictionaries and tuples) - here you just need to recognize a list when you see one.

## Module: NumPy in 10 minutes

You've met lists - Python's way of collecting things. But when it comes to **numbers**, Python has a superpower tool: **NumPy**. A numpy array looks like a list, but it can *do math*. See the difference:

````
import numpy as np

# A list "multiplied by 2" just becomes... two lists
regular_list = [1, 2, 3]
print(regular_list * 2)

# An array multiplied by 2 actually DOES MATH
array = np.array([1, 2, 3])
print(array * 2)
````

````
# np.arange(start, stop, step) - stop is NOT included
print(np.arange(0, 11))

# np.linspace(start, stop, count) - count evenly spaced numbers, stop IS included
print(np.linspace(0, 1, 5))
````

````
# Do math on a whole range of numbers in one go - e.g. VAT on all prices
prices = np.array([49, 129, 250, 999])
with_vat = prices * 1.25
print(with_vat)
````

## When code fails: Read the error message
You are going to make mistakes - everyone does, every day. The important thing is that Python's error messages are not scolding you, they are a pointer: they tell you *where* and *why*. Here is the error you will see most often. In Supplement B we practise more error types and a simple debugging tool.

````
# ⚠️ Expected error — this cell is supposed to fail
# We define a variable with the correct name
city_name = "Copenhagen"

# We try to print it, but misspell the variable name
print(city_nam)
````

#### How we read errors

Whoops! A red box. Let's break it down so it's not so scary.

```
---------------------------------------------------------------------------
NameError                                 Traceback (most recent call last)
<ipython-input-1-1a2b3c4d5e6f> in <module>
      3
      4 # We try to print it, but misspell the variable name
----> 5 print(city_nam)

NameError: name 'city_nam' is not defined
```

1.  **`Traceback (most recent call last)`**: This is just Python's way of saying: "Here's the trace of what went wrong, starting with the most recent event."
2.  **`----> 5 print(city_nam)`**: The arrow points to the exact line in your code where the error happened (line 5 in this case). This is incredibly useful!
3.  **`NameError: name 'city_nam' is not defined`**: This is the actual diagnosis. Let's translate it into plain English:
    *   **`NameError`**: The error type is a "Name Error".
    *   **`name 'city_nam' is not defined`**: Python looked in its memory for a box (variable) or a recipe (function) named `city_nam`, but couldn't find one.

**The solution:** We look at line 5, which the arrow points to. We see the name `city_nam`. Then we look at the rest of our code and see that we originally named our variable `city_name`. Ah! We're missing an 'e'. We fix the typo, and the code works.

````
# We define a variable with the correct name
city_name = "Copenhagen"

# We fix the typo
print(city_name)
````
