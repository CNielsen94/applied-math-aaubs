## Module 2: Ask Your Code Questions (Conditionals)

In the previous module, you learned how to store information in "named boxes", i.e. variables. Our programs can now remember things like names, cities, and numbers. But so far, our programs just run straight through, from top to bottom, executing every single instruction we give them.

But what if we want our program to make decisions? What if it should do one thing if it's raining, and another thing if the sun is shining? Or give a discount if a customer is over 65, but not if they're younger?

To do this, we need to learn how to ask our code questions. This is called **conditional logic**, and it's one of the most powerful ideas in all of programming.

### The Analogy: A Fork in the Road

Imagine you're driving down a road, and you come to a fork. You can't drive down both roads at once. You have to make a choice based on a condition. Maybe you ask yourself: "Is my destination to the left?" If the answer is "yes", you turn left. If the answer is "no", you continue straight ahead or turn right.

`if` statements in Python work exactly like a fork in the road. They let your program ask a yes/no question and choose a particular path (a block of code) to run, based on the answer.

### Yes or No: `True` and `False`

A computer doesn't understand complex answers. When we ask a question in our code, the answer must always boil down to a simple "yes" or "no". In programming, we call these two answers:

*   `True` (yes)
*   `False` (no)

These two values, `True` and `False`, are called **booleans**. They're fundamental to being able to make decisions. Let's see how we can ask questions that result in either `True` or `False`.

### Comparison operators: the tools for asking questions

To ask questions about our variables, we use **comparison operators**. They compare values and return either `True` or `False`.

Here are the most important ones:

| Operator | Meaning                  | Example        | Result |
| :------- | :-------------------------- | :-------------- | :------- |
| `==`     | Is equal to?                 | `5 == 5`        | `True`   |
| `!=`     | Is *not* equal to?          | `5 != 6`        | `True`   |
| `>`      | Is greater than?              | `10 > 5`        | `True`   |
| `<`      | Is less than?              | `10 < 5`        | `False`  |
| `>=`     | Is greater than or equal to? | `5 >= 5`        | `True`   |
| `<=`     | Is less than or equal to? | `10 <= 5`       | `False`  |

**Important note:** Notice `==` (two equals signs). This is a *comparison* ("is this equal to?"). A single equals sign, `=`, is an *assignment* ("put this value in the box"). Mixing these up is a very common beginner mistake!

#### Ask questions with operators

Let's create a couple of variables and ask them some questions. We'll print the result directly, so we can see whether the answer is `True` or `False`.

````
my_age = 30
your_age = 25

# Question 1: Is my age greater than your age?
print(f"Is my age greater than yours? Answer: {my_age > your_age}")

# Question 2: Are our ages the same?
print(f"Are our ages the same? Answer: {my_age == your_age}")

# Question 3: Is your age less than or equal to 25?
print(f"Is your age 25 or younger? Answer: {your_age <= 25}")
````

#### What happened?

Look at the output. Each `print` line asks a question, and the computer answers honestly with `True` or `False`.
1.  `my_age > your_age` becomes `30 > 25`, which is true. So the result is `True`.
2.  `my_age == your_age` becomes `30 == 25`, which is false. So the result is `False`.
3.  `your_age <= 25` becomes `25 <= 25`, which is true (because it's equal). So the result is `True`.

### Combining questions: logical operators

Sometimes one question isn't enough. What if we want to check whether a person is a teenager (i.e. older than 12 **and** younger than 20)? Here we need to combine two conditions. We do that with logical operators:

*   `and`: Both conditions must be `True` for the overall result to be `True`.
*   `or`: At least one of the conditions must be `True` for the overall result to be `True`.
*   `not`: Flips the result (`not True` becomes `False`).

#### Using `and` and `or`

````
age = 15
has_drivers_license = False

# Is the person a teenager? (age must be > 12 AND < 20)
is_teenager = age > 12 and age < 20
print(f"Is the person a teenager? {is_teenager}")

# Is the person allowed to drive? (must be at least 18 OR have a license)
# (This is a slightly silly example, but it shows the 'or' operator)
may_drive = age >= 18 or has_drivers_license == True
print(f"Is the person allowed to drive? {may_drive}")
````

#### What happened?
For `is_teenager`, `15 > 12` is `True` and `15 < 20` is also `True`. Because both sides of `and` are `True`, the overall result is `True`. For `may_drive`, `15 >= 18` is `False` and `has_drivers_license == True` is also `False`. Because neither side of `or` is `True`, the result is `False`.

Now that we know how to get a `True` or `False` answer, we can use it to steer our program down different paths.

### The `if-else` statement: two different paths

The `if` statement is good, but it only gives us one option: what to do if the condition is true. What if we want to do something else if it's false? Here we use an `else` statement.

The structure looks like this:

````
it_is_raining = False

if it_is_raining:
    print("Remember your umbrella when you go out!")

print("Have a good day!")
````

````
if condition:
    # Code that runs if the condition is True
else:
    # Code that runs if the condition is False
````

The program will *always* run either the `if` block or the `else` block, but never both.

### The `if-elif-else` chain: more than two paths

Sometimes we need more than just two paths. Imagine we need to assess the temperature: it could be "hot", "mild", or "cold". Here we have three possibilities.

For this, we use `elif`, which is a contraction of "else if". It lets us add extra conditions to our fork in the road.

The structure looks like this:

````
if condition_1:
    # Run this code if condition_1 is True
elif condition_2:
    # Run this code if condition_1 was False, BUT condition_2 is True
else:
    # Run this code if NONE of the above conditions were True
````

Python checks the conditions from top to bottom. As soon as it finds one that's `True`, it runs the corresponding code block and skips the rest of the chain.

#### I do: Grading

Let's make a simple program that gives feedback based on a point score.

````
score = 85

if score >= 90:
    print("Fantastic! You got top marks!")
elif score >= 70:
    print("Good job. A solid performance.")
elif score >= 50:
    print("You passed. Keep practicing.")
else:
    print("Unfortunately you didn't pass this time. Try again.")
````

#### Interpretation

Since `score` is 85, the following happens:
1.  Python checks `if score >= 90`. `85 >= 90` is `False`. It skips this block.
2.  Python moves on to `elif score >= 70`. `85 >= 70` is `True`. It runs this block and prints "Good job. A solid performance.".
3.  Because it found a true condition, it **completely ignores the rest of the `elif` and `else` blocks**, even though `85 >= 50` is technically also true. It stops at the first one it finds.

This is a very important point: the order of your `if` and `elif` conditions matters!

## Module 3: Repetition Makes Perfect (Loops)

In the previous modules, you learned to store data in variables and to make your program choose between options with `if-elif-else` statements. It's like giving your code a brain, so it can think and react.

But what if you need to perform the same action over and over? Imagine you have a list of 100 names, and you want to print a greeting to each person. Would you have to write 100 `print()` lines? That would be incredibly tedious and inefficient.

This is where **loops** come in. A loop is a way of telling the computer: "Run this block of code again and again, until a certain condition is met". It's one of the most time-saving and powerful tools a programmer has.

### The Analogy: An Assembly Line

Think of a loop as an **assembly line** in a factory.

*   You have a row of things to process (e.g. a box of Lego bricks).
*   You put all the things on the assembly line.
*   The assembly line brings one thing forward at a time to a work station.
*   At the work station, the exact same task is performed on each thing (e.g. "put a sticker on it").
*   When all the things from the assembly line have been processed, the work stops.

A `for` loop in Python works in exactly the same way. But before we can build our assembly line, we need something to put on it. That brings us to a new data type: **lists**.

### Lists: A Digital Shopping List

Until now, our "boxes" (variables) could only hold one thing at a time. A **list** is like a super-box that can hold a whole series of things in a particular order.

The best analogy is a **shopping list**.

*   It has a name (e.g. "my shopping list").
*   It contains several items (e.g. "milk", "bread", "cheese").
*   The order of the items matters.

In Python, we make a list by using square brackets `[]` and separating the individual elements with commas.

#### I do: Create and work with a list

Let's make a list of some Danish cities and see how we can interact with it.

````
for temp_variable in my_list:
    # Code that should run for each element.
    # This code is indented, just like with if statements.
    # 'temp_variable' will hold the current element from the list.
````

Let's break it down:
*   `for`: The keyword that starts the loop.
*   `temp_variable`: This is a name you choose yourself! On each pass through the loop, this variable will automatically hold the next element from the list. Think of it as the thing that's *currently* at the work station on the assembly line.
*   `in`: The keyword that connects the temporary variable to the list.
*   `my_list`: The list we want to go through.
*   `:`: The colon that starts the indented code block.

#### Greet the cities

Let's use our list `danish_cities` and a `for` loop to print a personal greeting to each city.

````
danish_cities = ["Copenhagen", "Aarhus", "Odense", "Aalborg"]

# Here our assembly line starts
# For each 'city' in the list 'danish_cities'...
for city in danish_cities:
    # ...perform this action
    print(f"Welcome to the fantastic city of: {city}!")

# our print() below only happens once all the cities in the block have been printed, because it is NOT indented
print("That was all the cities on the list.")

# You can try indenting print() into the block (press Tab twice),
# and see what happens to the output under the cell
````

#### What happened?
See how elegant that is! Instead of writing four `print()` lines, we wrote just one. The loop did the rest of the work for us.

Here's what happened, step by step:
1.  The loop starts. Python takes the first element from `danish_cities`, which is `"Copenhagen"`.
2.  It stores `"Copenhagen"` in the temporary variable `city`.
3.  It runs the indented code: `print(f"Welcome to the fantastic city of: {city}!")`. `city` is replaced with `"Copenhagen"`, and it prints "Welcome to the fantastic city of: Copenhagen!".
4.  The loop finishes its first pass. It goes back to the top and takes the next element, `"Aarhus"`.
5.  It stores `"Aarhus"` in `city` (overwriting the old value).
6.  It runs the indented code again, which now prints "Welcome to the fantastic city of: Aarhus!".
7.  This repeats for `"Odense"` and `"Aalborg"`.
8.  After `"Aalborg"` there are no more elements in the list. The assembly line is empty. The loop stops.
9.  The program continues running the code *after* the loop, and prints "That was all the cities on the list.".

### Repeating a set number of times: `range()`

What if we don't have a list, but just want to repeat an action a certain number of times? For that, Python has a super handy function called `range()`.

`range(5)` generates a sequence of numbers from 0 up to (but not including) 5. That is: 0, 1, 2, 3, 4. It's perfect to use in a `for` loop.

#### Using `range()`

````
# We want to print a message 3 times.
# 'i' is a traditional temporary variable for counters.
for i in range(3):
    print(f"This is repetition number {i+1}") # Indented block belongs to the loop

print("The loop is done.") # This block does NOT belong to the loop, since it isn't indented.
````

#### What happened?
The loop ran 3 times. The first time, `i` was `0`, the second time `1`, and the third time `2`. By writing `i+1` in our print statement, we got a nice counter starting from 1. `range()` is incredibly useful when you know exactly how many times something should repeat.

## Module 4: Write Your Own Recipes (Functions)

So far, we've written code that runs from top to bottom. We've learned to store data in variables, make choices with `if` statements, and repeat tasks with `for` loops. Our programs are getting more and more advanced!

But what happens when our programs get really big? Imagine a program with hundreds of lines of code. It can quickly become unmanageable. What if we have a particular calculation we need to perform in many different places in our program? Do we then copy and paste the same code again and again? That would be both cumbersome and a source of bugs. If we found an error in the calculation, we'd have to fix it everywhere.

The solution to this problem is one of the most elegant and fundamental ideas in programming: **functions**. A function is a named, reusable block of code that performs a specific task.

### The Analogy: A Recipe

The absolute best way to think about a function is as a **recipe**.

*   **The Recipe's Name:** A recipe has a name, e.g. "Pancake recipe". A function also has a name, e.g. `make_pancakes`.
*   **Ingredients (Parameters):** A recipe lists the ingredients you need (e.g. 2 eggs, 500 ml flour). These can vary - sometimes you'll make a small batch, other times a large one. In a function, these "ingredients" are called **parameters**. It's the information the function needs in order to do its job.
*   **Method (The Code Inside the Function):** The recipe has a series of steps you follow to turn the ingredients into a finished dish. This is the actual code inside the function.
*   **The Finished Dish (Return Value):** The result of following the recipe is a stack of pancakes. In a function, we can get a result back that we can keep working with. This is called a **return value**.
*   **Using the Recipe (Calling the Function):** When you decide to make pancakes, you *use* the recipe. In programming, we say we *call* the function. You can use the same recipe again and again, every time you feel like pancakes.

Why is this smart? Because you only need to write the recipe down once. After that, you (and others) can use it thousands of times without having to reinvent the wheel every time. This makes our code:
1.  **Reusable:** Write once, use many times.
2.  **Organized:** Break a big problem down into smaller, named "recipes" that are easy to understand.
3.  **Easier to maintain:** If you find a better way to make pancakes, you only need to fix it in one place: the recipe itself.

### I do: Defining and calling a simple function

Let's start with the simplest form of a function: one that doesn't take any ingredients and doesn't return a finished dish, but just performs an action.

The syntax for defining (writing) a function in Python looks like this:
*   We start with the keyword `def` (short for *define*).
*   Then comes the function's name (you choose it yourself, e.g. `print_a_greeting`).
*   After the name comes a set of parentheses `()`. This is normally where the ingredients (parameters) would go, but in this simple case they're empty.
*   We end the line with a colon `:`.
*   All the code belonging to the function must be **indented** (just like with `if` and `for`).

If you want to move on to Python's other collections - **dictionaries** and **tuples** - the supplement *Lists, dictionaries and tuples* is ready on the course page. It's especially useful no later than before Lesson 2.

````
# Step 1: We define our "recipe".
# Notice that the code isn't run yet. We've only written the recipe down.
def say_goodbye():
    print("That's all for now.")
    print("Thanks for coding along!")
    print("-------------------------")

# Step 2: Now we "use" the recipe by calling the function by its name.
print("The program is starting...")

say_goodbye() # Here we call the function

print("The program is now finished.")

say_goodbye() # We can call it again!
````

#### What happened?
Look at the output. The code inside `say_goodbye` only ran once we wrote `say_goodbye()`. Defining a function with `def` is like writing a recipe down in a cookbook - nothing happens until you actually open the book and decide to follow the recipe. We called the function twice, and both times the three `print` lines inside the function were executed.

### Adding Ingredients (Parameters)

A function only becomes really useful once we can give it information to work with. Let's make a function that can give a personal greeting. It needs one ingredient: the name of the person it should greet.

````
# Here 'name' is a parameter. It's a placeholder for the value
# we'll give the function when we call it.
def personal_greeting(name):
    print(f"Hello there, {name}! I hope you're having a fantastic day.")

# Now we call the function and give it an "argument".
# The argument "Mads" gets put into the parameter "name" inside the function.
personal_greeting("Mads")

# We can call it again with a different argument.
personal_greeting("Camilla")
````

#### What happened?
Here `name` is a **parameter** - a variable that only exists inside the function. When we call `personal_greeting("Mads")`, `"Mads"` is an **argument**. Python takes the argument `"Mads"` and copies it into the parameter `name`. Then the code inside the function runs with `name` set to `"Mads"`. Next time we call it with `"Camilla"`, the process repeats, but now with `name` set to `"Camilla"`. The function is the same, but the output changes based on the ingredients we give it.

#### Pro tip: Default ingredients (Default Parameters)
What if we sometimes forget to provide an ingredient? We can give a parameter a default value. Then it's used whenever we don't specify one ourselves.

````
def greeting_with_default(name="World"):
    print(f"Hi, {name}!")

greeting_with_default("Denmark") # Here we use the argument we provide
greeting_with_default()          # Here we give nothing, so the default value "World" is used
````

### Getting the finished dish back (Return Values)

Sometimes we don't want our function to print anything. Instead, we want it to calculate a value and give it back to us, so we can use it in the rest of our program. This is "the finished dish" from our recipe. For this, we use the keyword `return`.

#### A function that adds numbers together

Let's make a function that takes two numbers as parameters, adds them together, and *returns* the result.

````
def calculate_sum(num1, num2):
    total = num1 + num2
    return total # Stop the function and send the value of 'total' back

# We call the function and store the returned value in a new variable
result_from_function = calculate_sum(5, 3)

print(f"The result the function returned is: {result_from_function}")
print(f"We can now keep working with the result. E.g. multiply it by 2: {result_from_function * 2}")
````

#### Interpretation
When we called `calculate_sum(5, 3)`, the following happened:
1.  `num1` was set to `5`, and `num2` was set to `3` inside the function.
2.  `total` was calculated as `5 + 3 = 8`.
3.  `return total` sent the value `8` back to the place in the code where the function was called.
4.  The line `result_from_function = calculate_sum(5, 3)` effectively became `result_from_function = 8`.
5.  We could then print and continue working with the variable `result_from_function`.

#### A note on variable scope
A variable created *inside* a function (like `total` above) is **local**. This means it only exists while the function is running. Once the function finishes, the variable is "thrown out". You can't access it outside the function. This is smart, because it stops variables from different functions interfering with each other.

## Module 5: Debugging is your friend!

Congratulations on making it this far! You've learned about variables, conditionals, lists, loops, and functions. You now have a solid toolbox for building simple but powerful programs.

But along the way, you may have experienced something... frustrating. Maybe you forgot a colon, misspelled a variable name, or saw a big, red error message pop up. Your first reaction might have been: "Oh no, I've broken it!"

Let's establish one thing right away: **Errors aren't failures. Errors are feedback.**

Writing code is a conversation with the computer. Sometimes it misunderstands what you mean, and then it sends an error message back. This message isn't the computer scolding you; it's the computer trying to help you by saying: "I didn't quite understand this part. Could you look at line X and see if you meant something else?"

Even the most experienced programmers in the world see error messages every single day. The difference is that they've learned not to be afraid of them. They've learned to read them, understand them, and use them as a clue to find and fix the problem. The process of finding and fixing errors is called **debugging**.

In this module, we'll look at the most common error types you'll encounter as a new programmer. We'll learn to read the error messages and see them as a friendly guide instead of a roadblock.

### 1. `NameError`: "I don't know that name"

A `NameError` is probably the most common error of them all. It occurs when you ask Python to use a variable or function whose name it doesn't recognize. It's almost always caused by a simple typo.

#### A classic typo

Let's say we have a variable that stores a city. But when we go to print it, we misspell the name.

### 2. `SyntaxError`: "I don't understand your sentence"

A `SyntaxError` is different. Here, it's not a name Python can't find, but the actual *structure* or *grammar* of your code that's wrong. It's like writing an English sentence without a period, or with the words in completely the wrong order. You've broken the rules of the language.

These errors often happen because you forget a character like a colon `:`, a parenthesis `)`, or a quotation mark `"`.

#### I do: A missing colon

We know from Module 2 that an `if` statement always has to end with a colon. Let's see what happens if we forget it.

````
# ⚠️ Expected error — this cell is supposed to fail
temperature = 22

# We forget the colon ':' after the condition
if temperature > 20
    print("It's a hot day!")
````

This error message looks a bit different:
1.  **`SyntaxError: invalid syntax`**: The diagnosis is clear: "Syntax error: invalid grammar". Python is telling us that the way we've written the code doesn't follow the rules of the language.
2.  **`^`**: Notice the little arrow (caret). Python tries to be helpful and points to the place on the line where it got confused. It expected something more after `20` (namely a colon), but the line ended abruptly.

**The solution:** We see that the arrow points to the end of our `if` line. We remember the rule that `if`, `elif`, `else`, `for`, and `def` lines always have to end with a colon. We add it, and the code works.

### 3. `IndentationError`: "Your indentation is wrong"

This error is unique to Python and one of the most common for beginners. Because Python uses indentation (spaces at the start of a line) to define code blocks (what belongs to an `if`, `for`, `def`, etc.), it gets very confused if the indentation is unexpected.

### 4. `TypeError`: "You can't do that with that type of data"

A `TypeError` occurs when you try to perform an action on a data type that doesn't support that action. It's like trying to drink a chair or take a banana for a drive. The action ("drink") and the object ("chair") don't go together.

The most common `TypeError` for beginners is trying to add numbers and text strings together with `+`.

### A Simple Tool: `print()` Debugging

Sometimes your code runs without errors, but it doesn't do what you expect. Maybe the result of a calculation is wrong. How do you find out what's happening inside the program?

The simplest, but most effective, debugging technique is to use `print()` to inspect your variables at different stages.

````
# Suppose we have a bug in our logic
start_value = 10
factor = 5
result = (start_value + 22) * factor # Maybe it should have been + 2?

# We're not sure what 'result' is. Let's print it!
print(f"DEBUG: The value of 'result' is {result}")

if result > 50:
    print("The result is big!")
else:
    print("The result is small.")
````

By adding the `print()` statement, we can see that `result` is 160. If we expected it to be 60, we can now see that the error must lie in the calculation, and it's easier to see that `+ 22` should have been `+ 2`.
