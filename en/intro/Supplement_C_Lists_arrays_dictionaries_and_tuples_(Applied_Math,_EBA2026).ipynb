### Lists, Dictionaries, and Tuples in Python

#### **What is a List?**
A list in Python is an ordered collection of elements. Lists are very flexible and can contain different data types, including other lists.

##### **Why is this relevant in Python?**
Lists are among the most widely used data types in Python and are used in almost every kind of programming, from web development to scientific research.

##### **Why is it important to learn?**
Understanding lists is fundamental to manipulating data, building more complex data types, and understanding many other advanced features of Python.

#### **What is a Dictionary?**
A dictionary in Python is an unordered collection of key-value pairs. Here the key is unique and is used to access the corresponding value.

##### **Why is this relevant in Python?**
Dictionaries are extremely useful for storing data in a searchable way. For example, in a database-like structure.

##### **Why is it important to learn?**
Knowledge of dictionaries makes it easy to organize and manipulate data, especially when you are dealing with large amounts of data or complex data structures.

#### **What is a Tuple?**
A tuple is an ordered collection of elements, like a list, but it is immutable. This means that once it has been created, it cannot be changed.

##### **Why is this relevant in Python?**
Tuples are often used when you have a collection of data that should not change during the program's execution, such as configuration settings.

##### **Why is it important to learn?**
Understanding tuples helps you work with immutable data types and gives you a deeper understanding of how to manipulate data in Python safely.

#### The Basics of Python Functions and Methods

In Python there are both functions and methods, and it is important to understand the difference, as well as how they affect variables.

---

**Function vs Method**

1. **Function**: An independent routine that performs a specific task. It is a way of packaging code so it can be reused.

   Example: `len()`, `sorted()`

2. **Method**: A routine or function that is "bound" to a specific object and operates on that object.

   Example: `.append()`, `.remove()`

---

**In-place vs Return-based operations**

1. **In-place**: Some methods directly modify the object they operate on and do not return a new copy. This is known as "in-place" operations.

   🔍 **Example**: the `variable.append()` method adds an element to an existing list. This change happens "in-place", meaning the original list is modified, and no new list is returned.

2. **Return-based**: Other functions and methods return a new copy of the object with the changes, while the original object remains unchanged.

🔍 **Example**: the `sorted(variable)` function returns a new list that is a sorted copy of the original list, while the original list remains unchanged.

In the exercises that follow, you will be introduced to this way of thinking, and how to process/manipulate/access data using Python.

## About lists and arrays in Python via NumPy
Throughout the math course we have worked a lot with arrays and lists, and in many ways they resemble each other. There are, however, some differences you should be aware of when deciding whether to perform operations on lists or on arrays.

#### Lists vs Arrays - When Should You Use Which?

* Speed: Arrays in NumPy are more efficient to run operations on than lists.
* Functionality: NumPy comes with a range of built-in functions for numerical and statistical work.
* Types: Lists in Python can contain elements of different data types, while NumPy arrays are homogeneous.
* Size: Arrays generally take up less memory than lists.
* Complexity: Lists are simpler to use for general purposes and small tasks, while arrays are better for mathematical operations and larger-scale datasets.
* Built-in Methods: Lists have built-in methods like append(), remove(), etc., which can be more intuitive for general purposes.


Another note on lists vs. arrays is that lists are core Python (built in, can always be used), whereas arrays are part of the NumPy library (and therefore need to be imported)

#### Example of the Difference Between Lists and Arrays, Performance

* Use the time module to measure how long it takes to sum all the numbers in a very large list and a very large array. Is there a noticeable difference?

```python
import time
import numpy as np
```

```python
# List
big_list = list(range(1, 1000000))
start_time = time.time()
sum_list = sum(big_list)
end_time = time.time()
list_time = end_time - start_time
```

```python
# Array
big_array = np.arange(1, 1000000)
start_time = time.time()
sum_array = np.sum(big_array)
end_time = time.time()
array_time = end_time - start_time
```

```python
print(f"List Time: {list_time}, Array Time: {array_time}")
```

Lists are usually the easiest way to process and manipulate data, especially if you are working with different types of data, such as text, numbers, and objects at the same time. Arrays, on the other hand, are superior when it comes to computation and mathematical operations - consider which is best for the task you are trying to solve - and you can often easily switch between lists and arrays, depending on what you want to achieve.

#### What is a Dictionary?
A dictionary in Python is an unordered collection of data in a key:value pair format. Here are a few points explaining why dictionaries are important:

* Efficient Storage: Key-value pairs make it easy to store related data together.
*  Fast Access: Dictionaries give fast access to values based on unique keys, which makes them very efficient for lookups.
*  Flexibility: The values in a dictionary can be almost any Python type. So you can mix numbers, lists, other dictionaries, etc.
* Easy to Modify: You can easily add new key-value pairs or change values.
* Important in Other Data Structures: Dictionaries are often used in other data structures such as sets and JSON objects.

#### Task 1: Create Two Dictionaries

🔍 Explanation: We need two dictionaries, one containing the students' ages and one with grades. We use our student_grades from 2.1 and create a new dictionary with their ages (the names in your student_grades should match the following names:)

#### Task 2: Combine the Two Dictionaries into a New Dictionary

🔍 Explanation: The goal is to create a new dictionary that contains both sets of information. The key should still be the student's name, and the value should be another dictionary with the keys age and grade.

```python
# Empty dictionary to hold the combined data
combined_student_data = {}

# Combine the two dictionaries
for student, age in student_ages.items():
    grade = student_grades[student]  # we assume that every student also exists in `student_grades`
    combined_student_data[student] = {'age': age, 'grade': grade}
```

## Exercise 3: Tuples
In this exercise we will explore another important data type in Python called a 'tuple'. Tuples are very similar to lists, but they are immutable, meaning their contents cannot be changed after creation.


#### Task 1: Create a Tuple

🔍 Explanation: We start by creating a simple tuple representing a student's name, age, and grade.

```python
# Create a tuple
student_tuple = ('Peter', 21, 'A')
```

#### Task 2: Accessing Elements in a Tuple

🔍 Explanation: Even though tuples are immutable, you can still access their elements.

```python
# First element (name)
print(student_tuple[0])

# Second element (age)
print(student_tuple[1])

# Third element (grade)
print(student_tuple[2])
```

Task 3: Tuples in a List

🔍 Explanation: Tuples can be elements in a list. This is convenient for representing a collection of immutable data points.

```python
# Create a list of tuples
students = [('Peter', 21, 'A'), ('Anna', 22, 'A'), ('Mikkel', 23, 'B+')]

# Access the first tuple in the list
print(students[0])

# Access Peter's grade from the first tuple in the list
print(students[0][2])
```

#### Task 4: Immutability

🔍 Explanation: Demonstrate that tuples are immutable by trying to change a value.

```python
# Try to change a value (this will raise an error)
try:
    student_tuple[1] = 28
except TypeError as e:
    print(f"An error occurred: {e}")  # Prints: 'tuple' object does not support item assignment
```

## List comprehensions: the loop on a single line

Python has an elegant shorthand for when a loop is just building a list:

```python
# The long way
squares = []
for n in [1, 2, 3, 4, 5]:
    squares.append(n**2)
print(squares)

# The short way - one line, same result
squares2 = [n**2 for n in [1, 2, 3, 4, 5]]
print(squares2)
```

You don't need to use the short form yet - but notice how readable it is. It makes its full comeback in the course's Lesson 7.

## From API to numbers: Reading a data record
When you fetch data from the internet - for example electricity prices from an energy-data API - you typically get your data back as a list of **records**: one dictionary per measurement. That looks exactly like what you just learned:

```python
# One data record, as an API might return it
price_record = {"HourDK": "2026-01-15T17:00:00", "PriceArea": "DK1", "SpotPriceDKK": 892.5}

# Look up individual values by key
print(price_record["PriceArea"])
print(price_record["SpotPriceDKK"])
```

In practice you get a whole list of records - and then we combine three things you now know: a list, a dictionary lookup, and a NumPy array:

```python
import numpy as np

price_records = [
    {"HourDK": "2026-01-15T17:00:00", "PriceArea": "DK1", "SpotPriceDKK": 892.5},
    {"HourDK": "2026-01-15T18:00:00", "PriceArea": "DK1", "SpotPriceDKK": 1024.0},
    {"HourDK": "2026-01-15T19:00:00", "PriceArea": "DK1", "SpotPriceDKK": 731.25},
]

# Pull the prices out of each record and collect them in an array
prices = np.array([record["SpotPriceDKK"] for record in price_records])
print(prices)
print(f"Average price: {prices.mean():.2f} DKK/MWh")
```
