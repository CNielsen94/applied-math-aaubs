### Markdown in Notebooks
In a Jupyter Notebook you can use Markdown to format text in cells to explain what your code does, present the results, or write other kinds of explanatory text.

Here are some practical examples of how you can use Markdown in Jupyter Notebook:

**Headings:** You can create headings by using #. A single # makes a large heading, ## makes a smaller heading, ### makes an even smaller heading, and so on, up to six levels.

Example:

# This is a large heading
## This is a smaller heading
### This is an even smaller heading

**Bold or italic text:** You can make text bold by wrapping it in two asterisks or two underscores **bold text** or __bold text__. You can make text italic by wrapping it in one asterisk or one underscore *italic text* or _italic text_.

**Lists:** You can create bullet lists by using -, +, or *, and you can create numbered lists by using 1., 2., etc.

Example:
- This is
- a bullet list

1. This is
2. a numbered list

**Mathematical formulas:** Mathematical notation can be expressed by enclosing it in \$. For example: $y = ax + b$.

In Markdown, certain characters are reserved for formatting. These include characters such as #, *, _, [], (), >, !, $ and `. When Markdown sees these characters, it assumes you are trying to format your text in a specific way.

For example, if Markdown sees an asterisk (\*) or an underscore (\_), it may take that as a sign that you want to create italic or bold text. So if you write \*this\*, Markdown will convert it to "*this*" in italics.

If you actually want to write one of these characters without formatting your text, you need to use an "escape" character, which is the backslash (\\). Basically, the backslash (\\) allows you to use these special characters in your text without formatting it.

Markdown can do a lot, and I'd recommend that you either use my notebooks as a starting point, or make use of online resources, when you need to explain things yourselves in text format using markdown.

# How a Notebook Works

A notebook consists of cells that are run one at a time - and the order matters. When you run a cell (Shift+Enter or the ▶ button), Python remembers the result in the background, in what is called a **kernel**. Any later cell can use whatever the kernel has in memory.

The number in the square brackets to the left of a code cell, e.g. `[3]`, shows the order in which the cells were run. If the numbers don't increase from top to bottom, the cells have been run in a different order than the one they appear in - and that is the most common source of mysterious bugs in notebooks.

The two classic traps:

1. **You skip a cell.** If you run a cell that uses a variable from an earlier cell you never got around to running, it fails with a `NameError`.
2. **You delete code, but the kernel still remembers it.** If you create a variable, run the cell, and then delete the line afterward, the variable still exists in the kernel's memory. The notebook *looks* right, but only works until the kernel is restarted.

Let's try trap number 2 in practice:


````python
# Run this cell first
message = "I am remembered by the kernel"
````

````python
# Then run this cell - it works because 'message' is in the kernel's memory
print(message)
````

Now try the following experiment:

1. Delete the line `message = "I am remembered by the kernel"` in the cell above (or put a `#` in front of it), and run the cell again.
2. Run the `print(message)` cell again. It **still** works, even though the code that created the variable is gone - the kernel remembers it.
3. Now select **Runtime → Restart session and run all** (in Jupyter: *Kernel → Restart & Run All*). Now the cell fails with a `NameError`, because now the kernel's memory finally reflects the code that is actually in the notebook.

Hence the course's most important rule of thumb: **If something behaves strangely, restart the kernel and run all cells from top to bottom.** A notebook that cannot be run from top to bottom without errors is broken - even if it "worked" a moment ago.

# When the Code Fails: How to Read an Error Message

Errors are not the exception in programming - they are an everyday occurrence, even for experienced programmers. The difference is that experienced programmers have learned to *read* error messages instead of panicking over them.

When Python encounters an error, it prints what is called a **traceback**. Read it from the bottom up:

* The **bottom line** tells you the error type and what is wrong, e.g. `NameError: name 'the_number' is not defined`.
* The lines above show **where** it went wrong - an arrow (or a line number) points to the line that failed.

Below you'll find the three error types you will encounter most often. Run the cells and practice reading the error messages:


````python
# ⚠️ Expected error — this cell is supposed to fail
# NameError: you use a name the kernel doesn't know
# Usually a typo - or a cell that was never run
num_students = 200
print(num_studnets)
````

````python
# ⚠️ Expected error — this cell is supposed to fail
# SyntaxError: Python doesn't understand the statement
# Typically a missing colon, a missing parenthesis, or a missing quotation mark
x = 10
if x > 5
    print("x is greater than 5")
````

````python
# ⚠️ Expected error — this cell is supposed to fail
# TypeError: you mix types that don't go together - here text + number
num_tasks = 4
print("I have " + num_tasks + " assignments due")
````

### The Simplest Debugging Tool: print()

Sometimes the code doesn't fail at all - it just gives a wrong result. The trick then is to print along the way and check whether the variables actually contain what you think:


````python
price = 100
discount = 20
final_price = price - discount * 2   # the intention was: (price - discount) * 2

print(final_price)   # 60? We expected 160!
print(discount * 2)  # print the sub-calculation: 40. Aha - that's the missing parenthesis
````

There are more error types (`IndentationError`, `ValueError`, ...) and more advanced tools, but the approach is the same: read the bottom line, find the indicated line, and print your way to the problem. # NumPy and Matplotlib in Practice: A Trip Down Jomfru Ane Gade

Let's wrap up by combining variables, functions, and libraries in a (nearly) realistic example - and at the same time meet the two libraries you will use the most in this course.

Two bars on Jomfru Ane Gade, Aalborg's famous bar street, have different pricing models:

* **Bar 1** charges 50 kr. cover and 30 kr. per beer: $y = 30x + 50$
* **The dive bar** has free entry but charges 45 kr. per beer: $y = 45x$

Both are **linear functions**: the total price $y$ depends on the number of beers $x$. When does it pay off to pay the cover charge? We can both calculate and *see* our way to the answer.


````python
import numpy as np
import matplotlib.pyplot as plt

# np.arange(start, stop, step) creates a sequence of numbers - stop is NOT included, so we use 11 to reach up to 10
num_beers = np.arange(0, 11, 1)
print(num_beers)
````

````python
# You can do arithmetic on a numpy array all at once - no loops needed
price_bar1 = 30 * num_beers + 50
price_dive_bar = 45 * num_beers

print(price_bar1)
print(price_dive_bar)
````

````python
plt.plot(num_beers, price_bar1, 'o-', label="Bar 1 (50 kr. cover + 30 kr. per beer)")
plt.plot(num_beers, price_dive_bar, 'o-', label="The dive bar (45 kr. per beer)")
plt.title("What does the night out cost?")
plt.xlabel("Number of beers")
plt.ylabel("Total price (kr.)")
plt.legend()
plt.grid(True)
plt.show()
````

Read the graph: the lines cross between 3 and 4 beers. If you have fewer, the dive bar is cheapest - after that, the cover-charge bar pays off.

Notice what just happened: we translated an everyday situation into two linear functions, let numpy compute over entire number sequences at once, and let matplotlib draw the result. That is exactly the workflow the rest of the course builds on - in a moment we'll use it on mathematical functions, and already in the next lesson we will learn to find such intersection points precisely, instead of just reading them off the graph.
