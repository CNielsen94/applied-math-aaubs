### Eksponentialfunktionen

$f(x) = b^x$

En eksponentialfunktion er en matematisk funktion, hvor en variabel optræder i eksponenten. Den mest basale form for en eksponentialfunktion er $y = b^x$, hvor b er et positivt konstant tal forskellig fra 1. Hvis b er større end 1, vil funktionen vokse eksponentielt, mens hvis b er mellem 0 og 1, vil funktionen falde eksponentielt.

Eksponentialfunktioner er meget almindelige i mange videnskabelige, tekniske og økonomiske sammenhænge, fordi de ofte bruges til at beskrive processer, der vokser eller falder hurtigt og i en ikke-lineær skala, såsom befolkningstilvækst, radioaktivt henfald, eller rentesammensætning.

Her er nogle eksempler på, hvordan eksponentialfunktioner kan være relevante i erhvervsøkonomi:

**Renters rente:** Hvis du investerer penge i en bank eller en anden form for investering, og din investering vokser med en konstant rente, vil værdien af din investering vokse eksponentielt. Formlen for fremtidig værdi ved sammensat rente er $FV = P(1+\frac{r}{n})^{nt}$, hvilket er en form for eksponentialfunktion.

**Økonomisk vækst:** Når økonomer taler om langsigtede vækstrater, bruger de ofte eksponentielle modeller. Hvis en økonomi vokser med en konstant rate, vil dens størrelse over tid følge en eksponentialkurve.

**Afskrivninger:** Nogle typer af afskrivninger, såsom procentvis nedskrivning, kan modelleres med en eksponentiel funktion, hvor aktivets værdi falder eksponentielt over tid.

Derfor er det vigtigt at forstå, hvad en eksponentialfunktion er, og hvordan man arbejder med den. At kunne modelere og forstå eksponentiel vækst eller fald kan være afgørende for mange aspekter af erhvervsøkonomi.

### Logaritmen

Logaritmen er den inverse funktion til eksponentialfunktionen,
dvs. hvis $y = b^x$ , så er $x = \log_{b}(y)$

Logaritmer er et grundlæggende koncept inden for matematik og bruges meget i både videnskab og teknik. Logaritmer hjælper med at forenkle beregninger og gør det lettere at håndtere meget store eller meget små tal.

En logaritme angiver eksponenten, som en specifik base skal opløftes til, for at producere et tal. For eksempel, i udtrykket $10^2 = 100$, er 2 logaritmen for 100 i base 10. Vi kan skrive dette som $\log_{10}(100) = 2$.

**Forholdet mellem logaritmer og eksponenter:** Som vi har set, er en logaritme grundlæggende den omvendte operation af en eksponent. Det er nyttigt, fordi det gør det lettere at udføre beregninger, der ville være svære med bare eksponenter.

**Håndtering af store tal:** Logaritmer kan gøre meget store tal mere håndterbare. For eksempel, i stedet for at arbejde med en værdi som 1.000.000.000, kan vi arbejde med sin logaritme i base 10, som er 9 - dette kan skrives i python som $1e9$.

**Multiplikation og division bliver til addition og subtraktion:** En af de mest kraftfulde egenskaber ved logaritmer er, at de kan omdanne multiplikation og division til addition og subtraktion.

I det hele taget er logaritmer et kraftfuldt værktøj, der hjælper os med at forenkle og håndtere komplekse matematiske problemer.

## Praktiske eksempler

### Eksempel med compound interest (renters rente):

Funktionen for renters rente: $FV = P(1+\frac{r}{n})^{nt}$

$FV$ = Fremtidig værdi (Future value)

$P$ = Hovedstol (Principal)

$r$ = Rentesats (Interest rate)

$n$ = Antal gange renten forrentes pr. år (Number of interest compounds pr. year)

$t$ = Tid målt i år (Time, measured in years)



```python
# Definer variabler
P = 1000  # Hovedstol (det oprindelige lånebeløb eller investeringsbeløb)
r = 0.05  # Årlig rentesats (i decimalform, så 5% bliver 0.05)
n = 4     # Antal gange renten forrentes pr. år (vi antager kvartalsvis forrentning her)
t = 20    # Antal år investeringen løber
```

```python
# Liste til at gemme resultater
FV_list = []

# Beregn fremtidig værdi for hvert år og gem resultatet
for i in range(t+1):
    FV = P * (1 + r/n)**(n*i)
    FV_list.append(FV)
```

`range(n)` giver tallene $0, 1, ..., n-1$ - altså $n$ tal, der starter fra 0 og stopper FØR $n$ (en programmerings-quirk, der går igen overalt i Python).

I Python starter indeksering af lister og rækker fra 0. Derfor, når vi bruger **range(t+1)** i vores for-loop, sikrer vi, at loopet kører fra $i = 0$ til $i = t$, hvilket inkluderer $t$ selv. Hvis vi brugte **range(t)**, ville loopet kun køre fra $i = 0$ til $i = t-1$.

Her er hvordan loopet fungerer med hensyn til indeksering og beregning af $FV$ (Fremtidig Værdi):

* Når $i = 0$: $FV = P \times \left(1 + \frac{r}{n}\right)^{0} = P$ (Nuværende værdi)
* Når $i = 1$: $FV$ for det første år.
* Når $i = 2$: $FV$ for det andet år.
* Når $i = t$: $FV$ for det $t$'te år.

Hvis du brugte **range(t)** i stedet for **range(t+1)**, ville du ikke få $FV$ for det $t$'te år, da loopet ville stoppe ved $t-1$.

Så, $t+1$ i **range(t+1)** sikrer, at alle værdier fra år 0 til år $t$ er inkluderet i dine beregninger.


```python
FV_list
```

```python
# Print-forklaring
print(f"På 20 år har vi tjent {FV_list[20] - FV_list[0]:.2f}kr i renter. Der står altså {FV_list[20]:.2f}kr på vores konto efter en oprindelig investering på {int(FV_list[0])}kr.")
```

Bemærk: vi skrev `FV_list[20]` med et hardkodet indeks. Det virker kun, fordi `t = 20` - prøv at ændre `t` og se, hvad der sker. `FV_list[t]` ville følge med automatisk. Begge dele er "rigtige" - men kun den ene overlever, at nogen ændrer forudsætningerne.


```python
# Lav en figur
plt.figure()

# Plot resultaterne
plt.plot(range(t+1), FV_list)

# Tilføj titler og labels
plt.title("Vækst i investering over tid")
plt.xlabel("År")
plt.ylabel("Fremtidig værdi i kr.")

# Vis grafen
plt.show()
```

#### Find oprindelig indskud fra Fremtidig værdi

For identificere startindskuddet for en fremtidig værdi, kan følgende formel anvendes.

$P = \frac{FV}{\left(1 + \frac{r}{n}\right)^{nt}}$

```python
# Beregn P direkte
P = FV / (1 + r/n)**(n*t)

print(f"Den oprindelige investering var {P:.2f}.")
```

#### Find mængde af tid

Hvis t (antallet af år) er ukendt, kan vi bruge ligningen $FV = P * (1 + r/n)^{nt}$ til at finde t.

Vi isolerer t ved at omskrive denne ligning som:

$t = (log(FV/P) / log(1 + r/n)) / n$

Hvis vi anvender de oprindelige værdier for $P$, $r$ og $n$, og antager, at vi kender den fremtidige værdi ($FV$), kan vi finde $t$ med følgende Python-kode:

```python
# Definer kendte variabler
P = 1000  # Hovedstol
r = 0.05  # Årlig rentesats (decimal)
n = 4     # Antal gange renten forrentes pr. år
FV = 2701.4849407533275  # Fremtidige værdi

# Løs for t ved hjælp af logaritmeregneregler
t = (np.log(FV/P) / np.log(1 + r/n)) / n

print(f"Investeringen har løbet i {t:.2f} år.")
```

#### Identificer renten

$r = n \left( \left( \frac{FV}{P} \right)^{\frac{1}{nt}} - 1 \right)$

Den givne formel er designet til at beregne rentesatsen $r$ baseret på en kendt fremtidig værdi ($FV$), nutidsværdien af investeringen ($P$), antallet af rentebetalinger per periode ($n$), og antallet af perioder ($t$). Formlen er en omarrangering af den klassiske formel for fremtidig værdi i konteksten af sammensat rente.

Her er hvad der sker:

* $FV/P$ beregner multiplikatorfaktoren for investeringen.
* $(FV/P)^{1/(nt)}$ beregner roden af denne multiplikatorfaktor justeret for antallet af rentebetalinger og perioder.
*  $((FV/P)^{1/(nt)} - 1)$ trækker 1 fra denne rod for at isolere rentesatskomponenten.
* $n \cdot ((FV/P)^{1/(nt)} - 1)$ ganger til sidst med antallet af rentebetalinger per periode for at få den faktiske årlige rentesats $r$.

```python
# Antag at vi kender følgende værdier:
P = 1000  # Hovedstol
FV = 2701.4849407533275  # Fremtidsværdi
n = 4  # Antal forrentninger pr. år
t = 20  # Antal år
```

```python
# Beregn r ved hjælp af formlen:
r = n * ((FV/P)**(1/(n*t)) - 1)
```

```python
# rentesats som decimaltal
print("Rentesatsen er:", r)
```

```python
# rentesats som procent
print(f"Rentesatsen er: {r*100:.2f}%")
```

#### Logaritmen - fordobling af investering

Denne kode anvender den naturlige logaritme for at beregne, hvor lang tid det tager at fordoble en investering ved en given årlig rente $r$. Koden bruger den præcise formel - den såkaldte "Rule of 72" er en tommelfingerregel, der tilnærmer resultatet som $72$ divideret med renten i procent.

$t = \frac{\ln(2)}{\ln(1 + r)}$

Her er hvad der sker:

* $t$ = np.log(2) / np.log(1 + r): Dette bruger formelen for at finde $t$, antallet af år det tager at fordoble en investering. Formlen anvender naturlige logaritmer og er afledt af den generelle formel for sammensat rente. Det beregner $t$ ved at tage den naturlige logaritme af $2$ (fordi vi ønsker at fordoble investeringen) og dividere det med logaritmen af ($1+r$).

"Rule of 72" er en forenkling eller en tommelfingerregel, der kommer fra den mere præcise matematiske formel for sammensat rente. I den fuldstændige formel for sammensat rente bruger vi eksponenter til at udregne, hvor meget en investering vil vokse over tid.

For at finde tiden $t$, det tager for en investering at fordoble, sætter vi $FV=2×P$ i formlen for fremtidig værdi og løser for $t$. Dette giver os en ligning med en eksponentiel variabel, som vi kan løse ved at tage logaritmen på begge sider. I denne sammenhæng bruger vi den naturlige logaritme for sin matematiske enkelthed og fordi den er grundlaget for mange vækstprocesser i naturen og økonomi.

Så den naturlige logaritme bruges i den præcise beregning, fordi den gør det lettere at isolere $t$ og finde en løsning. "Rule of 72" er en forenkling af denne præcise matematiske model og bruger tallet 72 i stedet for at kræve, at man beregner logaritmer. Det er en nemmere, men mindre præcis, måde at estimere fordoblingstiden for en investering.

```python
# årlig rente
r = 0.05

# Beregn t ved hjælp af logaritmer
t = np.log(2) / np.log(1 + r)

print(f"Det tager {t:.2f} år at fordoble din investering.")

# Sammenlign med Rule of 72-tilnærmelsen
print(f"Rule of 72-estimatet giver: {72/(r*100):.2f} år.")
```

```python
# Beregn t ved hjælp af logaritmer
t = np.log(3) / np.log(1 + r)

print(f"Det tager {t:.2f} år at tredoble din investering.")
```

# Derivater
i matematik er "derivatet" og "den afledte" det samme. Begge termer refererer til den operation, der måler, hvor meget en funktion ændrer sig, når dens inputværdier ændres. Dette koncept er fundamentalt i calculus og bruges ofte til at bestemme hældningen af en funktion på et bestemt punkt, eller til at forudsige, hvordan en variabel vil ændre sig givet små ændringer i en anden variabel.

Mere formelt er den afledte af en funktion på et bestemt punkt defineret som grænsen for funktionens hældning mellem to punkter, når disse to punkter kommer uendeligt tæt på hinanden. Hvis denne grænse eksisterer, siges funktionen at være differentierbar på det punkt, og værdien af grænsen er funktionens afledte (eller derivat) på det punkt.

For eksempel, hvis vi har en funktion $f(x) = x^2$, så er dens afledte (eller derivat) $f'(x) = 2x$, hvilket betyder, at for hver lille ændring i x, vil $f(x)$ ændre sig med cirka $2x$ gange størrelsen af den ændring.

diff er en funktion i SymPy biblioteket, der bruges til at beregne derivatet af en funktion. Derivatet er en måling af, hvordan en funktion ændrer sig, når dens inputværdier ændrer sig. I praksis er det hældningen af en tangentlinje på et bestemt punkt på en kurve.

Diff kan bruges til forskellige ting, heriblandt:

* diff(f, x): Returnerer den første afledte af f med hensyn til x.
* diff(f, x, n): Returnerer den n'te afledte af f med hensyn til x.
* diff(f, x, y): Returnerer den første afledte af f med hensyn til x og derefter y (dvs. den blandet afledte).

For eksempel, hvis vi har en funktion $f = x^2 + 2x + 1$, kan vi beregne dens første afledte ved at skrive diff(f, x), hvilket vil give $2*x + 2$.

SymPy's diff funktion kan håndtere mange typer funktioner, herunder polynomier, trigonometriske, eksponentielle, logaritmiske, og sammensatte funktioner, og det kan anvendes med mange forskellige differentiationsregler, herunder produktreglen, kvotientreglen, og kædereglen

### Kædereglen
Kædereglen er en grundlæggende regel i differentialregning, der bruges til at finde den afledte af sammensatte funktioner. En sammensat funktion er en funktion, der er "bygget op" af to eller flere funktioner. For eksempel, hvis vi har en funktion $ f(x) $ og en anden funktion $ g(x) $, så er $ f(g(x)) $ en sammensat funktion.

Simpel sagt, kædereglen hjælper os med at bryde en sammensat funktion ned i dens "byggesten" og differentiere hvert stykke separat. Derefter multiplicerer vi disse afledte sammen for at få den samlede afledte.

Matematisk er kædereglen defineret som følger:
$\frac{d}{dx}[f(g(x))] = f'(g(x)) \times g'(x)$

Her repræsenterer $ f'(g(x)) $ den afledte af $ f $ evalueret i $ g(x) $, og $ g'(x) $ er den afledte af $ g $.


```python
# Vi importerer symbols, diff, exp, og cos fra sympy - Vi gør dette så vi kan kalde funktionerne direkte uden at skulle referere til sympy biblioteket
# Ved at køre denne kode, kan vi altså bruge symboler, differentiering, exponentialfunktionen, og cosinus fra SymPy pakken.
from sympy import symbols, diff, exp, cos, solve
```

#### Eksempel:

$f(x) = cos(x^2)$

Kædereglen er nødvendig her, fordi $f(x)$ er en sammensat funktion (⁡$cos$ af $x^2$). SymPy's .diff() metode opdager automatisk, hvornår den skal anvende kædereglen (samt andre differentiationsregler som produktreglen og kvotientreglen) afhængig af den funktion, du forsøger at differentiere. Dette gør det meget nemt at arbejde med komplekse eller sammensatte funktioner, da biblioteket håndterer alle de tekniske detaljer for os. Vi skal blot give det den funktion, vi ønsker at differentiere, og den variable med hensyn til hvilken vi vil tage den afledte, og .diff() klarer resten.

```python
# Definer x som symbol
x = symbols('x')

# Definer funktion, f
f = cos(x**2)
```

```python
# Udregn den afledte for cos(x^2)
derivative_f = f.diff(x)

#Vis den afledte
derivative_f
```

```python
# Find den afledte når x = .5
at_05 = derivative_f.subs(x, .5)
at_05
```

### Produktregel

Produktreglen er en metode i differentialregning til at finde den afledte af et produkt af to funktioner. Antag, at du har to funktioner $f(x)$ og $g(x)$. Produktreglen siger, at den afledte af deres produkt er:

\begin{equation}
(fg)'(x) = f'(x)g(x) + f(x)g'(x)
\end{equation}


Her repræsenterer $ f'(x) $ og $ g'(x) $ de afledte af $ f(x) $ og $ g(x) $ med hensyn til $ x $, og $ fg $ repræsenterer produktet af $ f $ og $ g $.

Med andre ord, for at finde den afledte af et produkt af to funktioner, tager du den første funktion og multiplicerer den med den afledte af den anden funktion, og tilføjer det til den anden funktion multipliceret med den afledte af den første funktion.

Produktreglen er meget nyttig i mange områder af matematik, fysik og ingeniørvidenskab, hvor du ofte skal finde den afledte af et produkt af to funktioner.


#### Eksempel:
$f(x) = exp(x)*cos(x)$

```python
#Definer x som symbol (bare for en sikkerheds skyld)
x = symbols('x')

# Definer funktion, f
f = exp(x)*cos(x)
```

```python
# Udregn den afledte
derivative_f = f.diff(x)
derivative_f
```

```python
# Find den afledte når x = .7
at_07 = derivative_f.subs(x, .7)
at_07
```

### Potensreglen (power rule)
Potensreglen er en metode i differentialregning til at finde den afledte af en funktion af formen $ f(x) = x^n $, hvor $ n $ er en konstant. Ifølge potensreglen er den afledte af $ f(x) $ givet ved:

\begin{equation}
f'(x) = n \times x^{(n-1)}
\end{equation}

Denne formel siger, at for at finde den afledte af $ x^n $, multiplicerer du eksponenten $ n $ med $ x $ opført til $ n-1 $. Potensreglen er meget anvendelig og udgør grundlaget for at finde den afledte af en lang række funktioner. Den er særlig nyttig, da den gør det muligt hurtigt at differentiere en bred klasse af funktioner, der indeholder potenser af $ x $.






#### Eksempel:

$f(x) = x^5$

```python
# Vi definerer igen x, bare for en sikkerheds skyld
x = symbols('x')

# Definer funktion, f
f = x**5

# Udregn den afledte
derivative_f = f.diff(x)
derivative_f
```

```python
derivative_f.subs(x, 5)
```

### Kvotient regel
Kvotientreglen er en metode i differentialregning, der anvendes til at finde den afledte af en kvotient af to funktioner, $ f(x) $ og $ g(x) $. Ifølge kvotientreglen er den afledte af kvotienten $ \frac{f(x)}{g(x)} $ givet ved:

\begin{equation}
\left( \frac{f(x)}{g(x)} \right)' = \frac{f'(x)g(x) - f(x)g'(x)}{(g(x))^2}
\end{equation}

Denne regel er specielt nyttig, når man arbejder med funktioner, der er delt af hinanden, og hvor man ønsker at finde den afledte af sådanne udtryk. Grundlæggende tager du den afledte af den øverste funktion og multiplicerer med den nederste funktion, og trækker derefter produktet af den øverste funktion og den afledte af den nederste funktion fra. Det hele skal herefter divideres med kvadratet af den nederste funktion.





#### Eksempel:

$f(x) = x^2$ og $g(x) = x + 1$

```python
x = symbols('x')
```

```python
# Definerer vores funktioner f og g
f = x**2
g = x + 1
```

```python
# Anvender kvotientreglen for at finde derivatet af h = f / g
h_prime = (g * diff(f, x) - f * diff(g, x)) / g**2
h_prime
```

```python
solve(h_prime)
```

Her anvender vi kvotientreglen til at finde den afledte af en kvotient af to funktioner $f(x)$ og $g(x)$.

Den afledte af $f(x)=x^2$ er $f′(x)=2x$, og den afledte af $g(x)=x+1$ er $g′(x)=1$.

Indsættes disse i kvotientreglen, får vi:

$$h' = \frac{(x+1) \cdot 2x - x^2 \cdot 1}{(x+1)^2}$$

solve($h′$) finder rødderne af den afledte funktion $h′$ hvilket vil give de værdier af $x$, hvor $h$ har lokale ekstrema (minimum eller maximum).

Rødderne $[−2,0]$ af den afledte funktion $h′$ indikerer de $x$-værdier, hvor funktionen $h(x)$ har lokale ekstrema. I praksis betyder det, at funktionen enten har et lokalt minimum eller maksimum på disse punkter, eller det kan være en "sadel"-punkt.

* x$=−2$ og $x=0$ er de værdier, hvor hældningen af $h(x)$ er 0.
* For at afgøre, om det er et minimum eller maksimum, kan man bruge andenafledte test eller blot kigge på funktionen på hver side af $x$-værdierne.

Så i denne kontekst, hvis vi ønsker at finde ud af mere om adfærden af $h(x)$ på disse punkter, kan du undersøge værdierne af $h(x)$ tæt på $x=−2$ og $x=0$. Det vil give dig en idé om, hvorvidt funktionen er stigende eller faldende nær disse punkter og dermed, om vi har at gøre med lokale minima eller maxima.
