* Indsæt værdier for a,b,c,x, og y.
* Indsæt ligningen for y1 og y2
* Kør funktionen (ved køre den har i skabt jeres egen funktion, som i kan bruge fremadrettet, til at tegne en sekant ud fra givne x-værdier på en bestemt ligning)
* Plot sekanten på ligningen ved x-værdierne 3 og 5
* Plot sekanten på ligningen ved x-værdierne -3 og -5
* Prøv eventuelt med andre x-værdier og se hvordan sekanten ændrer sig

```python
def draw_secant(x_values):
    # draw function
    a = DIT_UDTRYK_HER
    b = DIT_UDTRYK_HER
    c = DIT_UDTRYK_HER
    x = np.linspace(-20,30,100)
    y = DIT_UDTRYK_HER
    plt.plot(x,y)

    # hent x1 og x2
    x1 = x_values[0]
    x2 = x_values[1]

    # udregn y-værdier fra x1 og x2
    y1 = DIT_UDTRYK_HER
    y2 = DIT_UDTRYK_HER

    # hældning m og intercept udregnes her
    m = (y2-y1)/(x2-x1)
    intercept = y1-m*x1

    # formel for sekanten
    s = m*x+intercept

    # Plot sekantlinjen
    plt.plot(x,s, c='green')

    # Giv plottet en titel
    plt.title('Graf for Andengradsligning med Sekant')
    plt.grid(True)

    plt.show()
```
