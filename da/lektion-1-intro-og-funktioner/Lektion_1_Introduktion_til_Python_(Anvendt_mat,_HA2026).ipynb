### Markdown i Notebooks
I en Jupyter Notebook kan du bruge Markdown til at formatere tekst i celler for at forklare, hvad din kode gør, fremvise resultaterne, eller skrive andre former for forklarende tekst.

Her er nogle praktiske eksempler på, hvordan du kan bruge Markdown i Jupyter Notebook:

**Overskrifter:** Du kan lave overskrifter ved at bruge #. En enkelt # laver en stor overskrift, ## laver en mindre overskrift, ### laver endnu en mindre overskrift, og så fremdeles, op til seks niveau.

Eksempel:

# Dette er en stor overskrift
## Dette er en mindre overskrift
### Dette er endnu mindre overskrift

**Fed eller kursiv tekst:** Du kan gøre tekst fed ved at indramme den med to asterisker eller to understregninger **fed tekst** eller __fed tekst__. Du kan gøre tekst kursiv ved at indramme den med en asterisk eller en understregning *kursiv tekst* eller _kursiv tekst_.

**Lister:** Du kan lave punktlister ved at bruge -, +, eller *, og du kan lave nummererede lister ved at bruge 1., 2., osv.

Eksempel:
- Dette er
- en punktliste

1. Dette er
2. en nummereret liste

**Matematiske formler:** Matematisk notation kan udtrykkes ved at indsætte i \$. For eksempel: $y = ax + b$.

I Markdown, er visse tegn forbeholdt til formatering. Disse omfatter tegn som #, *, _, [], (), >, !, og $`. Når Markdown ser disse tegn, antager den, at du forsøger at formatere din tekst på en bestemt måde.

For eksempel, hvis Markdown ser en stjerne (\*) eller en understregning (\_), kan det tage det som et tegn på, at du ønsker at lave kursiv eller fed tekst. Så hvis du skriver \*dette\*, vil Markdown konvertere det til "*dette*" i kursiv.

Hvis du faktisk vil skrive et af disse tegn uden at formatere din tekst, skal du bruge et "escape" tegn, som er backslash (\\). Grundlæggende, backslash (\\) tillader dig at bruge disse specielle tegn i din tekst uden at formatere den.

Markdown kan mange ting, og jeg vil anbefale at i enten tager udgangspunkt i mine notebooks, eller benytter online resurcer, når i selv skal forklare ting i tekst-format vha. markdown.

# Sådan virker en notebook

En notebook består af celler, der køres én ad gangen - og rækkefølgen betyder noget. Når du kører en celle (Shift+Enter eller ▶-knappen), husker Python resultatet i baggrunden, i det der kaldes en **kernel**. Alle senere celler kan bruge det, kernen har i hukommelsen.

Tallet i den firkantede parentes til venstre for en kodecelle, fx `[3]`, viser i hvilken rækkefølge cellerne er blevet kørt. Hvis tallene ikke stiger oppefra og ned, er cellerne kørt i en anden rækkefølge end den, de står i - og det er den mest almindelige kilde til mystiske fejl i notebooks.

De to klassiske fælder:

1. **Du springer en celle over.** Kører du en celle, der bruger en variabel fra en tidligere celle, du aldrig fik kørt, fejler den med en `NameError`.
2. **Du sletter kode, men kernen husker den stadig.** Opretter du en variabel, kører cellen og sletter linjen bagefter, findes variablen stadig i kernens hukommelse. Notebooken *ser* rigtig ud, men virker kun indtil kernen genstartes.

Lad os prøve fælde nummer 2 af i praksis:


````python
# Kør denne celle først
besked = "jeg bliver husket af kernen"
````

````python
# Kør derefter denne celle - den virker, fordi 'besked' ligger i kernens hukommelse
print(besked)
````

Prøv nu følgende eksperiment:

1. Slet linjen `besked = "jeg bliver husket af kernen"` i cellen ovenfor (eller sæt `#` foran den), og kør cellen igen.
2. Kør `print(besked)`-cellen igen. Den virker **stadig**, selvom koden der oprettede variablen, er væk - kernen husker den.
3. Vælg nu **Runtime → Restart session and run all** (i Jupyter: *Kernel → Restart & Run All*). Nu fejler cellen med en `NameError`, for nu afspejler kernens hukommelse endelig den kode, der faktisk står i notebooken.

Heraf kursets vigtigste tommelfingerregel: **Opfører noget sig mærkeligt, så genstart kernen og kør alle celler oppefra og ned.** En notebook, der ikke kan køres fra top til bund uden fejl, er i stykker - også selvom den "virkede" for et øjeblik siden.

# Når koden fejler: Sådan læser du en fejlmeddelelse

Fejl er ikke undtagelsen i programmering - de er hverdagen, også for erfarne programmører. Forskellen er, at erfarne programmører har lært at *læse* fejlmeddelelser i stedet for at gå i panik over dem.

Når Python støder på en fejl, printer den en såkaldt **traceback**. Læs den nedefra og op:

* Den **nederste linje** fortæller fejltypen, og hvad der er galt, fx `NameError: name 'tallet' is not defined`.
* Linjerne ovenover viser **hvor** det gik galt - en pil (eller et linjenummer) peger på den linje, der fejlede.

Herunder ser I de tre fejltyper, I kommer til at møde oftest. Kør cellerne og øv jer i at læse fejlmeddelelserne:


````python
# ⚠️ Forventet fejl — denne celle skal fejle
# NameError: du bruger et navn, kernen ikke kender
# Oftest en stavefejl - eller en celle, der aldrig blev kørt
antal_studerende = 200
print(antal_studernede)
````

````python
# ⚠️ Forventet fejl — denne celle skal fejle
# SyntaxError: Python forstår ikke sætningen
# Typisk et manglende kolon, en manglende parentes eller et manglende citationstegn
x = 10
if x > 5
    print("x er større end 5")
````

````python
# ⚠️ Forventet fejl — denne celle skal fejle
# TypeError: du blander typer, der ikke passer sammen - her tekst + tal
antal_opgaver = 4
print("Jeg har " + antal_opgaver + " opgaver for")
````

### Det simpleste debugging-værktøj: print()

Nogle gange fejler koden slet ikke - den giver bare et forkert resultat. Så er tricket at printe undervejs og tjekke, om variablerne faktisk indeholder det, du tror:


````python
pris = 100
rabat = 20
slutpris = pris - rabat * 2   # meningen var: (pris - rabat) * 2

print(slutpris)   # 60? Vi forventede 160!
print(rabat * 2)  # print delberegningen: 40. Aha - det er parentesen, der mangler
````

Der findes flere fejltyper (`IndentationError`, `ValueError`, ...) og mere avancerede værktøjer, men fremgangsmåden er den samme: læs nederste linje, find den udpegede linje, og print dig frem til problemet. # NumPy og Matplotlib i praksis: En tur i Jomfru Ane Gade

Lad os slutte af med at kombinere variabler, funktioner og biblioteker i et (næsten) realistisk eksempel - og samtidig møde de to biblioteker, I kommer til at bruge mest i dette kursus.

To barer i Jomfru Ane Gade har forskellige prismodeller:

* **Bar 1** tager 50 kr. i entré og 30 kr. pr. øl: $y = 30x + 50$
* **Bodegaen** har gratis entré, men tager 45 kr. pr. øl: $y = 45x$

Begge er **lineære funktioner**: den samlede pris $y$ afhænger af antallet af øl $x$. Hvornår kan det betale sig at betale entré? Det kan vi både regne og *se* os frem til.


````python
import numpy as np
import matplotlib.pyplot as plt

# np.arange(start, stop, trin) laver en række af tal - stop er IKKE inkluderet, så vi bruger 11 for at nå op til 10
antal_øl = np.arange(0, 11, 1)
print(antal_øl)
````

````python
# Et numpy-array kan der regnes på i ét hug - ingen løkker nødvendige
pris_bar1 = 30 * antal_øl + 50
pris_bodega = 45 * antal_øl

print(pris_bar1)
print(pris_bodega)
````

````python
plt.plot(antal_øl, pris_bar1, 'o-', label="Bar 1 (50 kr. entré + 30 kr. pr. øl)")
plt.plot(antal_øl, pris_bodega, 'o-', label="Bodegaen (45 kr. pr. øl)")
plt.title("Hvad koster byturen?")
plt.xlabel("Antal øl")
plt.ylabel("Samlet pris (kr.)")
plt.legend()
plt.grid(True)
plt.show()
````

Aflæs grafen: linjerne krydser mellem 3 og 4 øl. Bliver det til færre, er bodegaen billigst - derefter er det entré-baren, der kan betale sig.
