### Lister, Dictionaries, og Tuples i Python

#### **Hvad er en Liste?**
En liste i Python er en ordnet samling af elementer. Lister er meget fleksible og kan indeholde forskellige datatyper, inklusiv andre lister.

##### **Hvorfor er det relevant i Python?**
Lister er blandt de mest anvendte datatyper i Python og bruges i næsten alle former for programmering fra webudvikling til videnskabelig forskning.

##### **Hvorfor er det vigtigt at lære?**
Forståelse af lister er grundlæggende for at kunne manipulere data, bygge mere komplekse datatyper, og forstå mange andre avancerede funktioner i Python.

#### **Hvad er en Dictionary?**
En dictionary i Python er en uordnet samling af nøgle-værdi-par. Her er nøglen unik og bruges til at tilgå den tilsvarende værdi.

##### **Hvorfor er det relevant i Python?**
Dictionaries er ekstremt nyttige til at lagre data i en søgbar måde. For eksempel, i en database-lignende struktur.

##### **Hvorfor er det vigtigt at lære?**
Kendskab til dictionaries gør det nemt at organisere og manipulere data, især når I har at gøre med store mængder data eller komplekse datastrukturer.

#### **Hvad er en Tuple?**
En tuple er en ordnet samling af elementer, ligesom en liste, men den er uforanderlig. Dette betyder, at når den først er oprettet, kan den ikke ændres.

##### **Hvorfor er det relevant i Python?**
Tuples bruges ofte, når I har en samling af data, der ikke bør ændres gennem programmets kørsel, som f.eks. konfigurationsindstillinger.

##### **Hvorfor er det vigtigt at lære?**
At forstå tuples hjælper med at arbejde med uforanderlige datatyper og giver et dybere forståelse for, hvordan man kan manipulere data i Python på en sikker måde.

#### Grundlæggende om Python Funktioner og Metoder

I Python findes der både funktioner og metoder, og det er vigtigt at forstå forskellen, samt hvordan de påvirker variabler.

---

**Funktion vs Metode**

1. **Funktion**: En uafhængig rutine, der udfører en bestemt opgave. Det er en måde at pakke kode på, så den kan genbruges.
   
   Eksempel: `len()`, `sorted()`

2. **Metode**: En rutine eller funktion, der er "bundet" til et bestemt objekt og optræder på det objekt.

   Eksempel: `.append()`, `.remove()`

---

**In-place vs Return-baserede operationer**

1. **In-place**: Nogle metoder ændrer direkte det objekt, de opererer på, og returnerer ikke en ny kopi. Dette er kendt som "in-place" operationer.
   
   🔍 **Eksempel**: `variabel.append()` metoden tilføjer et element til en eksisterende liste. Denne ændring sker "in-place", hvilket betyder, at den oprindelige liste ændres, og der returneres ikke en ny liste.

2. **Return-baseret**: Andre funktioner og metoder returnerer en ny kopi af objektet med ændringerne, mens det oprindelige objekt forbliver uændret.

🔍 **Eksempel**: `sorted(variabel)` funktionen returnerer en ny liste, der er en sorteret kopi af den originale liste, mens den originale liste forbliver uændret.

I de følgende øvelser, vil i blive introduceret for denne form for tænkning, og hvordan man kan bearbejde/manipulere/tilgå data vha. Python.

## Om lister og arrays i Python via NumPy
I gennem matematik-undervisningen har vi arbejdet meget med arrays og lister, og de minder på mange måder om hinanden. Der er dog nogle forskelligheder man skal være opmærksom på, når man overvejer om man skal lave operationer på lister eller på arrays.

#### Lister vs Arrays - Hvornår skal man bruge hvad?

* Hurtighed: Arrays i NumPy er mere effektive at køre operationer på end lister.
* Funktionalitet: NumPy kommer med en række indbyggede funktioner for numerisk og statistisk arbejde.
* Typer: Lister i Python kan indeholde elementer af forskellige datatyper, mens NumPy arrays er homogene.
* Størrelse: Arrays optager generelt mindre hukommelse end lister.
* Kompleksitet: Lister er enklere at bruge til generelle formål og små opgaver, mens arrays er bedre til matematiske operationer og datasæt i større skala.
* Indbyggede Metoder: Lister har indbyggede metoder som append(), remove(), osv., der kan være mere intuitiv for generelle formål.


En anden note om lister vs. arrays, er at lister er grundlæggende Python (indbygget, kan altid bruges), hvorimod arrays er en del af NumPy biblioteket (og skal derfor importeres)

#### Eksempel på forskel mellem lister og arrays, ydeevne

* Brug time modulet til at måle, hvor lang tid det tager at summere alle tal i en meget stor liste og en meget stor array. Er der en bemærkelsesværdig forskel?

```python
import time
import numpy as np
```

```python
# Liste
big_list = list(range(1, 1000000))
start_time = time.time()
sum_list = sum(big_list)
end_time = time.time()
list_time = end_time - start_time
```

```python
# Array
big_array = np.arange(1, 1000000)
start_time = time.time()
sum_array = np.sum(big_array)
end_time = time.time()
array_time = end_time - start_time
```

```python
print(f"List Time: {list_time}, Array Time: {array_time}")
```

Lister er oftest nemmest til at bearbejde og manipulere med data, specielt hvis man arbejder med forskellige typer af data, som for eksempel tekst, tal, og objekter på samme tid. Arrays er derimod overlegne når det kommer til udregning og matematiske operationer - overvej hvornår hvad er bedst til den opgave i søger at løse - og man kan ofte nemt skifte mellem lister og arrays, alt efter hvad man vil opnå.

## Øvelse 2 - Dictionaries (ordbøger)
Dictionaries er en af de indbyggede datastrukturer i Python, der bruges til at opbevare en samling af elementer. Disse elementer er gemt som en række nøgle-værdi-par.

#### Hvad er en Dictionary?
En dictionary i Python er en uordnet samling af data i et nøgle:værdi par-format. Her er nogle punkter, der forklarer, hvorfor dictionaries er vigtige:

* Effektiv Lagring: Nøgle-værdi-par gør det nemt at gemme relaterede data sammen.
*  Hurtig Adgang: Dictionaries giver hurtig adgang til værdier baseret på unikke nøgler, hvilket gør dem meget effektive til søgning.
*  Fleksibilitet: Værdierne i en dictionary kan være næsten hvilken som helst Python-type. Så man kan blande tal, lister, andre dictionaries osv.
* Let at ændre: Du kan tilføje nye nøgle-værdi-par eller ændre værdier nemt.
* Vigtige i andre Datastrukturer: Dictionaries bruges ofte i andre datastrukturer som sæt og JSON-objekter.

## Øvelse 3: Tuples
I denne øvelse vil vi udforske en anden vigtig datatype i Python kaldet 'tuple'. Tuples er meget lig med lister, men er uforanderlige, hvilket betyder, at deres indhold ikke kan ændres efter oprettelsen.



#### Opgave 1: Opret en Tuple

🔍 Forklaring: Vi starter med at oprette en simpel tuple, der repræsenterer en students navn, alder og karakter.

```python
# Opret en tuple
student_tuple = ('Peter', 21, 'A')
```

#### Opgave 2: Adgang til elementer i en Tuple

🔍 Forklaring: Selvom tuples er uforanderlige, kan I stadig tilgå deres elementer.

```python
# Første element (navn)
print(student_tuple[0])

# Andet element (alder)
print(student_tuple[1])

# Tredje element (karakter)
print(student_tuple[2])
```

Opgave 3: Tuples i en Liste

🔍 Forklaring: Tuples kan være elementer i en liste. Dette er praktisk for at repræsentere en samling af uforanderlige datapunkter.

```python
# Opret en liste af tuples
students = [('Peter', 21, 'A'), ('Anna', 22, 'A'), ('Mikkel', 23, 'B+')]

# Adgang til den første tuple i listen
print(students[0])

# Adgang til Peter's karakter fra den første tuple i listen
print(students[0][2])
```

#### Opgave 4: Uforanderlighed

🔍 Forklaring: Demonstrer, at tuples er uforanderlige, ved at forsøge at ændre en værdi.

```python
# Forsøg at ændre en værdi (dette vil give en fejl)
try:
    student_tuple[1] = 28
except TypeError as e:
    print(f"An error occurred: {e}")  # Outputter: 'tuple' object does not support item assignment
```

## List comprehensions: løkken på én linje

Python har en elegant kortform, når en løkke bare skal bygge en liste:


```python
# Den lange vej
kvadrater = []
for n in [1, 2, 3, 4, 5]:
    kvadrater.append(n**2)
print(kvadrater)

# Den korte vej - én linje, samme resultat
kvadrater2 = [n**2 for n in [1, 2, 3, 4, 5]]
print(kvadrater2)
```

Når man henter data fra internettet - fx elpriser fra Energi Data Service, som kursets gennemgående case gør - får man typisk sine data tilbage som en liste af **records**: én dictionary per måling. Det ligner præcis det, du lige har lært:

I praksis får man en hel liste af records - og så kombinerer vi tre ting, du nu kan: en liste, et dictionary-opslag og et NumPy-array:

```python
import numpy as np

elpris_records = [
    {"HourDK": "2026-01-15T17:00:00", "PriceArea": "DK1", "SpotPriceDKK": 892.5},
    {"HourDK": "2026-01-15T18:00:00", "PriceArea": "DK1", "SpotPriceDKK": 1024.0},
    {"HourDK": "2026-01-15T19:00:00", "PriceArea": "DK1", "SpotPriceDKK": 731.25},
]

# Træk priserne ud af hver record og saml dem i et array
priser = np.array([record["SpotPriceDKK"] for record in elpris_records])
print(priser)
print(f"Gennemsnitspris: {priser.mean():.2f} DKK/MWh")
```
