### Eksempel: $f(x,y)=x^2+y^2$

I dette eksempel ser vi på funktionen $f(x,y)=x^2+y^2$. Denne funktion repræsenterer en paraboloid. Vi starter med at visualisere denne funktion.

```python
# De sædvanlige pakker
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, diff, solve

# Importer redskab til at lave 3d visualisering i matplotlib
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Definer funktionen
def f(x, y):
    return x**2 + y**2
```

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
```

```python
X, Y = np.meshgrid(x, y)
Z = f(X, Y)
```

```python
# Plot funktionen
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

plt.show()
```

### Partielle Afledede

Når vi differentierer en funktion af flere variable, skal vi tage højde for hver variabel separat. Dette giver os de såkaldte "partielle afledede".

Partielle afledede beskriver, hvordan funktionen ændrer sig, når vi ændrer en af de uafhængige variable, mens vi holder den anden konstant.

```python
# Definer x og y som symboler
x, y = symbols('x y')

# Definer funktionen, f
f = x**2 + y**2
```

```python
f_x = diff(f, x)  # differentierer med hensyn til x
f_y = diff(f, y)  # differentierer med hensyn til y
```

```python
print('Første ordens partielle afledede:')
print('df/dx:', f_x)
print('df/dy:', f_y)
```

### Kritiske Punkter

Kritiske punkter er de punkter, hvor alle de første ordens partielle afledede er lig med nul eller ikke eksisterer. Disse punkter er interessante, fordi de ofte repræsenterer lokale minima, lokale maksima eller saddelpunkter.

```python
# Find de kritiske punkter ved at sætte de partielle afledede lig med 0
kritiske_punkter = solve([f_x, f_y], (x, y))

print('\nKritiske punkter:', kritiske_punkter)
```

I dette eksempel er $f(x, y) = x^2 + y^2$ en funktion af to variable. Vi visualiserer denne funktion ved at bruge matplotlib, hvilket giver os en tredimensional overfladeplot.

Derefter differentierer vi funktionen f med hensyn til både x og y ved hjælp af sympy's diff funktion. Dette giver os de første ordens partielle afledede af funktionen.

Endelig, for at finde de kritiske punkter i funktionen (dvs., steder hvor gradienten er 0), løser vi for x og y når de partielle afledede er lig med 0. I dette tilfælde er det kritiske punkt ved (0, 0), hvilket er minimumspunktet for funktionen.

## Komplekse funktioner

I dette eksempel ser vi på en mere kompleks funktion $ f(x, y) = -2x^2 - 3y^2 + xy + 4y + 50 $. Funktionen visualiseres igen ved hjælp af en 3D overfladeplot.

```python
# Definer funktionen
def f(x, y):
    return -2*x**2 - 3*y**2 + x*y + 4*y + 50
```

```python
# Definer x og y variable
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
```

```python
# Definer X og Y som et meshgrid.
X, Y = np.meshgrid(x, y)
```

```python
# Definer Z som en funktion af X og Y
Z = f(X, Y)
```

```python
# Plot funktionen
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

plt.show()
```

### Første-ordens partielle afledede
For at analysere denne funktion nærmere, finder vi de første-ordens partielle afledede med hensyn til $ x $ og $ y $.

Disse afledede repræsenterer stigningen af funktionen $ f $ i retningerne $ x $ og $ y $.


```python
# Lad os differentiere funktionen
x, y = symbols('x y')
f = -2*x**2 - 3*y**2 + x*y + 4*y + 50
```

```python
f_x = diff(f, x)  # differentierer med hensyn til x
f_y = diff(f, y)  # differentierer med hensyn til y
```

```python
print('Første ordens partielle afledede:')
print('df/dx:', f_x)
print('df/dy:', f_y)
```

### Kritiske punkter
Vi finder kritiske punkter for funktionen ved at løse de ligninger, hvor de første-ordens partielle afledede er lig med 0:

\begin{equation}
\text{{Kritiske punkter: løsning til }} \frac{\partial f}{\partial x} = 0 \text{{ og }} \frac{\partial f}{\partial y} = 0
\end{equation}

```python
# Find de kritiske punkter ved at sætte de partielle afledede lig med 0
kritiske_punkter = solve([f_x, f_y], (x, y))
```

```python
print('\nKritiske punkter:', kritiske_punkter)
```

### Anden-ordens partielle afledede
For at bestemme arten (maksimum, minimum eller saddelpunkt) af de kritiske punkter, finder vi de anden-ordens partielle afledede:

\begin{equation}
\frac{\partial^2 f}{\partial x^2} = \text{{d}}^2f/\text{{dx}}^2, \quad \frac{\partial^2 f}{\partial y^2} = \text{{d}}^2f/\text{{dy}}^2, \quad \frac{\partial^2 f}{\partial x \partial y} = \text{{d}}^2f/\text{{dxdy}}
\end{equation}

#### Om de anden-ordens partielle afledede $ f_{xy} $ og $ f_{yx} $

I matematik er det generelt sådan, at hvis en funktion er tilstrækkeligt "glat" (kontinuert og differentiabel), så vil de blandede anden-ordens partielle afledede være lig hinanden. Dette er kendt som **Schwarz' teorem**, og det kan formuleres som:

\begin{equation}
\frac{\partial^2 f}{\partial x \partial y} = \frac{\partial^2 f}{\partial y \partial x}
\end{equation}

Hvis dette er tilfældet, og $ f_{xy} $ allerede er beregnet, så er det ikke nødvendigt at beregne $ f_{yx} $, da de vil være ens. Derfor kan $ f_{yx} $ i mange tilfælde anses som overflødig i beregningerne, især hvis vi er sikre på, at funktionen opfylder betingelserne for Schwarz' teorem.

I praksis beregnes begge værdier ofte som en form for dobbelttjek, for at sikre at funktionen faktisk opfylder disse betingelser. Men hvis $ f_{yx} $ ikke bruges i yderligere beregninger, kan den anses for overflødig.


```python
# anden-ordens partielle afledede
f_xx = diff(f_x, x)
f_yy = diff(f_y, y)
f_xy = diff(f_x, y)
f_yx = diff(f_y, x)
```

```python
print('Anden ordens partielle afledede:')
print('d^2f/dx^2:', f_xx)
print('d^2f/dy^2:', f_yy)
print('d^2f/dxdy:', f_xy)
print('d^2f/dydx:', f_yx)
```

### Hessian Determinant
Vi kan nu beregne Hessian-determinanten, $ D $, som er givet ved:

\begin{equation}
D = \frac{\partial^2 f}{\partial x^2} \frac{\partial^2 f}{\partial y^2} - \left( \frac{\partial^2 f}{\partial x \partial y} \right)^2
\end{equation}

```python
# Hessian determinant, D = f_xx*f_yy - f_xy^2
D = f_xx*f_yy - f_xy**2
print(f"Determinanten af Hessian matricen er {D}")
```

### Test af kritiske punkter
Endelig kan vi benytte Hessian-determinanten og de anden-ordens partielle afledede til at afgøre arten af de kritiske punkter:

* Hvis $ D > 0 $ og $ \frac{\partial^2 f}{\partial x^2} > 0 $ (eller $ \frac{\partial^2 f}{\partial y^2} > 0 $), har vi et lokalt minimum.
* Hvis $ D > 0 $ og $ \frac{\partial^2 f}{\partial x^2} < 0 $ (eller $ \frac{\partial^2 f}{\partial y^2} < 0 $), har vi et lokalt maksimum.
* Hvis $ D < 0 $, har vi et saddelpunkt.
* Hvis $ D = 0 $, er testen ikke afgørende.

```python
# Test det kritiske punkt
D_val = D.subs(kritiske_punkter)
f_xx_val = f_xx.subs(kritiske_punkter)

if D_val > 0 and f_xx_val > 0:
    print(f"Punktet {kritiske_punkter} er et lokalt minimum.")
elif D_val > 0 and f_xx_val < 0:
    print(f"Punktet {kritiske_punkter} er et lokalt maksimum.")
elif D_val < 0:
    print(f"Punktet {kritiske_punkter} er et saddelpunkt.")
else:
    print("Testen er ikke afgørende for punktet", kritiske_punkter)
```

### Forklaring om `.subs()` Funktionen i Sympy og Dens Rolle i Udregningen

I vores kode bruger vi `.subs()` metoden fra Sympy-biblioteket. Formålet med denne metode er at erstatte symbolske variabler med konkrete tal. I vores tilfælde bruger vi det til at substituere de kritiske punkter (x, y) ind i de andenordens partielle afledede (`f_xx`) og Hessian determinanten (`D`).

Måske har du bemærket, at værdierne af `f_xx` og `D` ikke ændrer sig efter substitutionen. Dette skyldes, at vi i dette eksempel har arbejdet med en funktion, hvor de partielle afledede og deres determinant ikke er afhængige af de kritiske punkter. I generelle tilfælde vil værdierne dog ændre sig, og det er derfor vigtigt at udføre denne substitution for præcist at bestemme, om et kritisk punkt er et minimum, maksimum eller saddelpunkt.

I nogle tilfælde kan den andenordens partielle afledede og Hessian determinanten ($D$) være funktioner af de variable ($x$, $y$). I sådanne tilfælde er det nødvendigt at bruge .subs() for at evaluere deres værdier ved de kritiske punkter.

Lad os se et eksempel:

Vi overvejer en funktion $f(x,y)=x^3-3xy+y^3$.

```python
# Definer variabler og funktion
x, y = symbols('x y')
f = x**3 - 3*x*y + y**3
```

```python
# Førsteordens partielle afledede
f_x = diff(f, x)
f_y = diff(f, y)
```

```python
# Find de kritiske punkter
kritiske_punkter = solve([f_x, f_y], (x, y))
```

```python
kritiske_punkter
```

#### Reelle vs. komplekse løsninger

Bemærk outputtet ovenfor: `solve` fandt fire løsninger, men to af dem indeholder $i = \sqrt{-1}$ - de er **komplekse tal**. Vores funktion lever i det reelle $(x, y)$-plan, så kun de **reelle** løsninger er faktiske kritiske punkter. Dem filtrerer vi fra med et list comprehension, inden vi klassificerer:


```python
# Behold kun de reelle løsninger - kun de er kritiske punkter for funktionen
reelle_punkter = [p for p in kritiske_punkter if all(koord.evalf().is_real for koord in p)]
reelle_punkter
```

```python
# Andenordens partielle afledede
f_xx = diff(f_x, x)
f_yy = diff(f_y, y)
f_xy = diff(f_x, y)
```

```python
# Hessian determinant, D
D = f_xx*f_yy - f_xy**2
```

```python
f_xx
```

```python
# Evaluér D og f_xx ved de reelle kritiske punkter
for punkt in reelle_punkter:
    # Pak værdierne i en dictionary
    subs_dict = {x: punkt[0], y: punkt[1]}

    D_val = D.subs(subs_dict)
    f_xx_val = f_xx.subs(subs_dict)

    if D_val > 0 and f_xx_val > 0:
        print(f"Punktet {punkt} er et lokalt minimum.")
    elif D_val > 0 and f_xx_val < 0:
        print(f"Punktet {punkt} er et lokalt maksimum.")
    elif D_val < 0:
        print(f"Punktet {punkt} er et saddelpunkt.")
    else:
        print(f"Testen er ikke afgørende for punktet {punkt}")
```

```python
# Definer funktionen
def f(x, y):
    return x**3 - 3*x*y + y**3
```

```python
# Definer x og y variable
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
```

```python
# Definer X og Y som et meshgrid.
X, Y = np.meshgrid(x, y)
```

```python
# Definer Z som en funktion af X og Y
Z = f(X, Y)
```

```python
# De to reelle kritiske punkter
x1, y1, z1 = 0, 0, f(0, 0)
x2, y2, z2 = 1, 1, f(1, 1)
```

```python
# Plot funktionen
fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6)
ax.scatter(x1, y1, z1, c='r', marker='o', s=100, label='Kritisk punkt (0, 0) - saddelpunkt')
ax.scatter(x2, y2, z2, c='g', marker='o', s=100, label='Kritisk punkt (1, 1) - lokalt minimum')

ax.legend()

plt.show()
```

## Eksempel med løsning af funktion med mange variable (ikke pensum)
Denne del er ikke en del af pensum, men efter at have forberedt undervisning, blev jeg nysgerrig på, hvordan løsning af systemer med mange ubekendte variable ville se ud. Jo flere ubekendte, jo mere komplekst bliver udregningerne, og derfor bliver det også sværere at bruge de biblioteker vi har anvendt ind til videre. I dette eksempel benyttes Sympy, hvilket er brugbart op til omkring 4 ubekendte (eller flere, hvis man er tålmodig). Hvis i er interesserede i at arbejde videre med funktioner med mange ubekendte, er Python biblioteket SciPy udviklet til komplekse matematiske udregninger og løsning af store systemer.

Det bliver også mere kompliceret at få en intuitiv fornemmelse af, om man laver de korrekte steps, da det ikke er muligt at visualisere systemer med mange variable.

```python
from sympy import symbols, diff, solve, Matrix

# Definer variable og funktion
x1, x2, x3, x4 = symbols('x1 x2 x3 x4')
f = x1**2 + x2**2 - x3**2 - x4**2
```

```python
# Beregn de partielle afledede
partial_x1 = diff(f, x1)
partial_x2 = diff(f, x2)
partial_x3 = diff(f, x3)
partial_x4 = diff(f, x4)
```

```python
# Find de kritiske punkter ved at løse ligningssystemet
critical_points = solve([partial_x1, partial_x2, partial_x3, partial_x4], (x1, x2, x3, x4))
```

```python
# Beregn Hessian-matricen
hessian = Matrix([[diff(partial_x1, x1), diff(partial_x1, x2), diff(partial_x1, x3), diff(partial_x1, x4)],
                  [diff(partial_x2, x1), diff(partial_x2, x2), diff(partial_x2, x3), diff(partial_x2, x4)],
                  [diff(partial_x3, x1), diff(partial_x3, x2), diff(partial_x3, x3), diff(partial_x3, x4)],
                  [diff(partial_x4, x1), diff(partial_x4, x2), diff(partial_x4, x3), diff(partial_x4, x4)]])
```

```python
hessian
```

```python
# Print kritiske punkter
print("Kritiske punkter: ", critical_points)
```

```python
critical_points
```

```python
# Beregn Hessian-matrix ved det kritiske punkt
hessian_at_point = hessian.subs(critical_points)
```

```python
hessian_at_point
```

```python
# Beregn egenværdier
eigen_values = hessian_at_point.eigenvals()
```

```python
# Print egenværdier
print(f"Egenværdier ved punktet {critical_points}: {eigen_values}")
```

```python
# Klassificer det kritiske punkt
if all(val > 0 for val in eigen_values):
    print(f"Punktet {critical_points} er et lokalt minimum.")
elif all(val < 0 for val in eigen_values):
    print(f"Punktet {critical_points} er et lokalt maksimum.")
else:
    print(f"Punktet {critical_points} er et saddelpunkt.")
```

## Forklaring: Hvorfor bruger vi egenværdier til at klassificere kritiske punkter?

Når vi arbejder med funktioner af flere variable, f.eks. $ f(x_1, x_2, \ldots, x_n) $, er en af vores opgaver at identificere og klassificere kritiske punkter. Kritiske punkter er steder, hvor alle førsteordens partielle afledede er lig med nul. Disse punkter er af interesse, fordi de ofte repræsenterer lokale minimums- eller maksimumsværdier, eller andre interessante egenskaber ved funktionen.

### Hvad er Hessian-matricen?

For at klassificere disse kritiske punkter bruger vi ofte Hessian-matricen. Denne $ n \times n $ matrix er en samling af funktionens andenordens partielle afledede. Hessian-matricen giver os vigtige informationer om, hvordan funktionen "bøjer" omkring et givet punkt.

### Hvorfor er egenværdier vigtige?

Egenværdierne af Hessian-matricen giver os specifikke indikationer om denne "bøjning":

- **Alle positive egenværdier**: Hvis alle egenværdier er positive, har vi et "skål"-formet område omkring det kritiske punkt. Dette indikerer, at det kritiske punkt er et **lokalt minimum**.

- **Alle negative egenværdier**: Omvendt, hvis alle egenværdier er negative, har vi et "kuppel"-formet område. Dette indikerer, at det kritiske punkt er et **lokalt maksimum**.

- **Blandede egenværdier**: Hvis vi har både positive og negative egenværdier, har funktionen en "saddel"-form i nærheden af det kritiske punkt. Dette betyder, at det kritiske punkt er et **saddelpunkt**, som hverken er et minimum eller maksimum.

### Sammenfatning

Egenværdierne af Hessian-matricen giver os altså en metode til geometrisk at forstå og klassificere kritiske punkter i funktioner af flere variable. De hjælper os med at afgøre, om et kritisk punkt er et lokalt minimum, et lokalt maksimum eller et saddelpunkt.
