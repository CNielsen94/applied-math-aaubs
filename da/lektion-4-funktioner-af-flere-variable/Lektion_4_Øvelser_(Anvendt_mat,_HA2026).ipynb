#### Øvelse 3.1 - Formulering af problemet
* Formuler en omkostningsfunktion $ C(x, y) = x^2 + 2y^2 $, hvor $ x $ og $ y $ repræsenterer mængderne af to forskellige råmaterialer.
* Definer en begrænsning for råmaterialerne som **constraint** = $ 4x + 2y - 100 $.



#### Kort om Lagrange-metoden

Lagrange-metoden bruges, når vi vil minimere (eller maksimere) en funktion, *mens* en betingelse skal overholdes - her: minimér omkostningerne $C(x, y)$, samtidig med at produktionen opfylder $4x + 2y = 100$.

Tricket er at samle det hele i én ny funktion, **Lagrange-funktionen**:

$$L(x, y, \lambda) = C(x, y) - \lambda \cdot (4x + 2y - 100)$$

Den nye variabel $\lambda$ ("lambda") holder styr på betingelsen: parentesen er nemlig 0, præcis når betingelsen er opfyldt. Herefter gør vi, som vi plejer med kritiske punkter - finder de partielle afledte af $L$ (nu med hensyn til alle tre variable) og sætter dem lig 0. Løsningen er det punkt, der giver de lavest mulige omkostninger uden at bryde betingelsen.

I økonomi har $\lambda$ oven i købet sin egen fortolkning: den viser cirka, hvor meget omkostningerne ændrer sig, hvis betingelsen lempes med én enhed (fx hvis produktionskravet hæves fra 100 til 101). Den fortolkning møder I igen senere på studiet.
