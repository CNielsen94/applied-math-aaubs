## Differentiering Recap
Differentieringsregler bruges i calculus (matematik) til at finde derivaterne af forskellige funktioner. At differentiere en funktion betyder at finde en ny funktion, der beskriver stigningen i den oprindelige funktion.

Et derivat (den afledte) er et matematisk koncept, der kommer fra calculus, en gren af matematik, der beskæftiger sig med ændring og bevægelse.

Derivatet af en funktion repræsenterer hastigheden for ændringen af funktionen ved et bestemt punkt. Det kan tænkes på som stigningen af tangentlinjen til funktionen ved dette punkt. Hvis funktionen repræsenterer en position i forhold til tid, så vil dens derivat repræsentere hastigheden.

Matematisk kan det derivat af en funktion ved et punkt beregnes som grænseværdien af ændringen i funktionens værdi delt med ændringen i input, som denne ændring i input går mod nul.

Funktionen, der associerer hvert punkt med sit derivat, kaldes for den afledte funktion eller simpelthen derivatet af den oprindelige funktion.

For eksempel, hvis vi har en funktion $f(x) = x^2$, så er derivatet af denne funktion $2x$. Det betyder, at for ethvert punkt $x$, er stigningen af tangentlinjen til kurven ved dette punkt $2x$.

Derivater spiller en nøgle rolle indenfor mange områder af videnskaben og teknikken, herunder fysik, ingeniørarbejde, økonomi, og mere, fordi de tillader os at kvantificere og forstå ændring.

# Optimering

Optimering er processen med at finde det bedst mulige resultat givet en bestemt funktion og nogle begrænsninger. I matematik og økonomi er det ofte et spørgsmål om at maksimere eller minimere en given funktion.

For eksempel kan en virksomhed forsøge at maksimere sin profit givet en bestemt produktion, eller en ingeniør kan forsøge at minimere materialeanvendelsen i en konstruktion uden at gå på kompromis med dens styrke.

Differentiering bruges i denne sammenhæng til at finde lokale maksimum og minimum af funktioner, det vil sige punkter, hvor funktionen skifter fra at stige til at falde, eller omvendt. Det sker ved at finde punkterne, hvor funktionens derivat (eller stigningstallet) er nul eller ikke eksisterer. Disse punkter er kandidater for lokale ekstrema (dvs. maksimum eller minimum).

Lad os tage et eksempel. Antag at en virksomhed har en profitfunktion $P(x) = -2x^2 + 20x - 30$, hvor x repræsenterer antallet af enheder produceret og solgt. Virksomheden ønsker at finde det antal enheder, der maksimerer profitten.

Først differentierer vi profitfunktionen for at finde dens derivat: $P'(x) = 20 -4x$.

Dernæst sætter vi denne afledede funktion lig med nul og løser for $x: -4x + 20 = 0$ giver $x = 5$.

Dette punkt er et kritisk punkt for profitfunktionen, og ved at analysere den afledte funktion (eller ved at anvende anden-derivat-testen, hvis vi har en funktion, der er differentierbar mere end én gang), kan vi afgøre, om dette punkt er et maksimum eller minimum. I dette tilfælde, da den afledede funktion falder for værdier større end 5, er $x = 5$ et maksimum for profitfunktionen.

Så ved at bruge differentiering har vi fundet, at virksomheden maksimerer sin profit ved at producere og sælge 5 enheder.

Som vi har set tidligere, kan dette også løses i Python. I det følgende eksempel vil vi differentiere profitfunktionen, sætte den lig med nul, og løse for x:

* Vi starter med at importerer nødvendige pakker, og definere $x$ som symbol samt vores funktion, $P$.

```python
from sympy import symbols, diff, solve
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
```

```python
# Definér variablen
x = symbols('x')
```

```python
# Definér profitfunktionen
P = -2*x**2 + 20*x - 30
```

I SymPy, når vi bruger **solve** funktionen på en ligning, vil den forsøge at finde alle de værdier af den ukendte variabel (i dette tilfælde $x$) som opfylder ligningen.

Hvis vi har en funktion $f(x)$ og dens afledte $f′(x)$, så vil **solve(f_prime, x)** give de værdier af $x$, hvor $f′(x)=0$. Disse er de kritiske punkter for funktionen, hvor den enten har et lokalt minimum, et lokalt maksimum, eller et vendepunkt.

I matematisk analyse og calculus refererer et "kritisk punkt" på en funktion til et punkt, hvor den første derivat $ f'(x) $ enten er nul ($ f'(x) = 0 $) eller ikke eksisterer. Det er i disse punkter, at funktionen har potentiale for at have et lokalt maksimum, et lokalt minimum, eller en vandret tangent (dvs., et "fladt" sted på kurven).

**Hvorfor løser vi $f'(x) = 0$**:

Når vi differentierer en funktion $ f(x) $, får vi dens første derivat $ f'(x) $, som repræsenterer hældningen af tangenten til kurven i et bestemt punkt $ x $. Hvis $ f'(x) = 0 $, betyder det, at tangenten til kurven i dette punkt er vandret. Dette indikerer en mulig lokal maksimal eller minimal værdi af $ f(x) $.

For eksempel, hvis vi har en funktion $ f(x) = x^2 $, vil den første derivat være $ f'(x) = 2x $. Når vi løser $ f'(x) = 0 $, får vi $ x = 0 $, hvilket er det kritiske punkt. Hvis vi plotter $ f(x) = x^2 $, vil vi se, at det er i dette punkt, funktionen har sit minimum.

Dette gøres i Python vha. disse steps

* **Differentier funktionen**: Først tager man den første derivat af funktionen $ f(x) $, hvilket giver $ f'(x) $.
* **Løs for kritiske punkter**: Derefter løser man $ f'(x) = 0 $ for at finde de værdier af $ x $, hvor $ f'(x) $ er nul. Disse $ x $-værdier er de kritiske punkter.
* (I nogle situationer) **Tjek anden derivat**: For at finde ud af, om de kritiske punkter er lokale minima eller maxima, kan man bruge den anden derivat $ f''(x) $. Hvis $ f''(x) > 0 $ i et kritisk punkt, har vi et lokalt minimum. Hvis $ f''(x) < 0 $, har vi et lokalt maksimum.

Så, når vi ser udtrykket "Løs $ f'(x) = 0 $", handler det om at finde disse kritiske punkter, hvor funktionen potentielt har et lokalt maksimum eller minimum.

med udgangspunkt i vores eksempelfunktion $P(x) = -2x^2 + 20x - 30$ kan vi nu følge disse steps:

```python
# Find derivatet (den afledte)
P_prime = diff(P, x)
```

```python
P_prime
```

```python
# Find x-værdi(er), hvor afledte er 0 (kritisk punkt(er))
critical_points = solve(P_prime, x)
```

```python
print(f"Kritiske punkter: {critical_points}")
```

```python
# Find anden afledte
P_double_prime = diff(P_prime, x)
```

```python
P_double_prime
```

```python
# Output som symbol (i SymPy format)
test = P_double_prime.subs(x, 5)
```

```python
test
```

```python
# Evaluer andet derivat i det kritiske punkt - med output som numerisk værdi (med evalf())
second_derivative_test = P_double_prime.evalf(subs={x: critical_points[0]})
```

```python
second_derivative_test
```

**evalf()** er en metode i SymPy, der bruges til at evaluere symbolske udtryk numerisk. Når vi bruger **evalf()** med **subs={x: critical_points[0]}**, evaluerer vi det andet derivat af profitfunktionen $P(x)$ ved det første kritiske punkt, som er blevet fundet tidligere i koden.

**evalf()** gør det muligt at tage et symbolsk udtryk og evaluere det numerisk ved en given værdi af variablen. I dette tilfælde bruges det til at udføre anden-derivat-testen ved det kritiske punkt, som vi har fundet. I dette tilfælde er det ikke strengt taget nødvendigt, men der kan være situationer hvor derivat-testen eller anden derivat-testen bliver vist som en symbolsk værdi (f.eks derivatet af eksponentialfunktionen) - for at sikre at at det kan aflæses af numpy og matplotlib.pyplot bruger vi evalf() til at sikre numerisk output.

Vi kan også få Python til at udskrive denne test for os (Hvis man for eksempel har mange, eller vil automatisere output til læsning). I dette eksempel bruger vi en else/if (elif) betingele - dette ville, for eksempel, kunne kobles sammen med et for-loop, hvis man havde mange kritiske punkter til test.

```python
# If/else betingelse, som testes:
if second_derivative_test < 0:
    print(f"x={critical_points[0]} er et maksimum.")
elif second_derivative_test > 0:
    print(f"x={critical_points[0]} er et minimum.")
else:
    print("Anden-derivat-testen er ikke afgørende.")
```

elif er en forkortelse for "else if" og er en del af en **if-elif-else** kontrolstruktur i Python. Denne struktur bruges til at teste flere betingelser i rækkefølge og udføre en kodeblok, når den første sande betingelse er fundet.

I ovenstående eksempel ser vi denne if-elif-else struktur anvendt til at udføre anden-derivat-testen:

* **If**-blokken: if second_derivative_test < 0: tester, om værdien af det andet derivat er mindre end 0. Hvis dette er sandt, udføres den tilhørende kodeblok (**print(f"x={critical_points[0]} er et maksimum.")**), og resten af strukturen ignoreres.
* **Elif**-blokken: elif second_derivative_test > 0: tester, om værdien af det andet derivat er større end 0, men kun hvis den foregående if-betingelse ikke var sand. Hvis denne elif-betingelse er sand, udføres den tilhørende kodeblok (**print(f"x={critical_points[0]} er et minimum.")**), og resten af strukturen ignoreres.
* **Else**-blokken: Hvis ingen af de ovenstående betingelser er sande, udføres koden i else-blokken (**print("Anden-derivat-testen er ikke afgørende.")**).

Så for at opsummere:

* **if** tester den første betingelse.
* Hvis **if**-betingelsen er falsk, går det videre til **elif** og tester denne betingelse.
* Hvis begge **if** og **elif** er falske, udføres **else**-blokken.

Kun én af disse blokke vil udføres, afhængigt af hvilken betingelse der er sand først. Dette er en effektiv måde at håndtere flere betingelser på i en organiseret og letforståelig måde.

En funktion kan teoretisk set have et uendeligt antal kritiske punkter, afhængigt af funktionens natur og kompleksitet. Et kritisk punkt er et sted, hvor en funktions første derivat er nul eller ikke eksisterer.

For en simpel polynomisk funktion af første grad, som $f(x) = ax + b$, er der ingen kritiske punkter, da dens derivat er konstant (og ikke nul). For en polynomisk funktion af anden grad, som $f(x) = ax^2 + bx + c$, er der et enkelt kritisk punkt, som er toppunktet eller bundpunktet for parabolen.

Polynomiske funktioner af højere grad kan have flere kritiske punkter. For eksempel kan en kubisk funktion, som $f(x) = ax^3 + bx^2 + cx + d$, have op til to kritiske punkter.

Ikke-polynomiske funktioner, som trigonometriske, eksponentielle, logaritmiske og andre komplekse funktioner, kan have et uendeligt antal kritiske punkter. For eksempel, funktionen $f(x) = sin(x)$ har et uendeligt antal maksima og minima, der ligger ved hver top og bund af sinus-bølgen.


### Udregning og visualisering af funktioner med mange kritiske punkter

I den følgende kode fokuserer vi på at identificere og visualisere lokale minima og maksima i en funktion med et uendeligt antal løsninger. Koden handler om at finde og visualisere kritiske punkter (lokale maksima og minima) på en sinus-bølge. Her er en simpel forklaring:

Dette kunne være en fin måde at illustrere, hvordan virksomheder kunne have forskellige "høj- og lavpunkter" over tid, hvis deres indtjening følger en lignende bølgeformet mønster.

```python
# Find lokale maksima og minima ved at benytte sig af find_peaks fra scipy library
local_max_indices, smid_væk = find_peaks(y)
local_min_indices, smid_væk = find_peaks(-y)
```

```python
from sympy import symbols, diff, solve, sin

# Definér symboler og funktion
x = symbols('x')
f = sin(x)
```

```python
# Find første og andet derivat
f_prime = diff(f, x)
f_double_prime = diff(f_prime, x)
```

```python
f_prime
```

```python
f_double_prime
```

```python
# Find kritiske punkter
critical_points = solve(f_prime, x)
```

```python
critical_points
```

```python
# Evaluer andet derivat i de kritiske punkter
second_derivative_evaluated = [f_double_prime.evalf(subs={x: point}) for point in critical_points]

print("Kritiske punkter: ", critical_points)
print("Andet derivat evalueret i de kritiske punkter: ", second_derivative_evaluated)
```

```python
critical_points[0].evalf()
```

### Kritiske punkter: $[ \frac{\pi}{2}, \frac{3\pi}{2} ]$

Dette output indikerer, at der er kritiske punkter ved $ x = \frac{\pi}{2} $ og $ x = \frac{3\pi}{2} $ for funktionen $ f(x) = \sin(x) $ i én periode af sinusbølgen. Disse punkter svarer til steder, hvor den første afledte $ f'(x) = \cos(x) $ er lig med 0.

### Andet derivat evalueret i de kritiske punkter: $[-1, 1]$

Dette viser værdien af det andet derivat $ f''(x) $ ved de kritiske punkter. Værdien af det andet derivat kan bruges til at afgøre, om et kritisk punkt er et minimum, maksimum, eller saddelpunkt:

- $ f''(x) > 0 $ : Lokalt minimum
- $ f''(x) < 0 $ : Lokalt maksimum
- $ f''(x) = 0 $ : Ukendt (saddelpunkt eller fladt område)

For $ f(x) = \sin(x) $:

- Ved $ x = \frac{\pi}{2} $, er $ f''(x) = -1 $, hvilket indikerer et lokalt maksimum.
- Ved $ x = \frac{3\pi}{2} $, er $ f''(x) = 1 $, hvilket indikerer et lokalt minimum.

Så disse værdier angiver, at der er et lokalt maksimum ved $ \frac{\pi}{2} $ og et lokalt minimum ved $ \frac{3\pi}{2} $ for én periode af sinusbølgen. Disse punkter vil gentage sig i hver periode af funktionen.


## Kritisk punkt uden lokalt minimum eller maksimum
Husk, at tilstedeværelsen af kritiske punkter ikke nødvendigvis betyder, at der er lokale maksima eller minima. For at bestemme dette, skal teknikker som anden-derivat-testen bruges.

Et godt eksempel på et kritisk punkt, der hverken er et lokalt minimum eller maksimum, er et saddelpunkt.

Tag for eksempel funktionen $f(x) = x^3$. Hvis vi tager derivatet af denne funktion, får vi $f'(x) = 3x^2$. Derivatet er 0, når $x = 0$, så $x = 0$ er et kritisk punkt for funktionen.

Men ser vi på funktionens opførsel tæt på 0, er $f(x)$ faktisk voksende på *begge* sider af $x = 0$ — den afledte $f'(x) = 3x^2$ er aldrig negativ. Så i stedet for et lokalt minimum eller maksimum har vi ved $x = 0$ et punkt, hvor hældningen rammer nul og derefter vokser igen, uden en klar 'dal' eller 'top'. Formelt kaldes sådan et punkt i funktioner af én variabel et **stationært vendepunkt**; her bruger vi også den bredere betegnelse **saddelpunkt**, som I måske kender.

Vi kan visualisere det med følgende pythonkode:

```python
# x som symbol
x = symbols('x')

#funktion
f = x**3
```

```python
# Find den første og anden afledte af f
f_prime = diff(f, x)
f_double_prime = diff(f_prime, x)
```

```python
f_prime
```

```python
f_double_prime
```

```python
# Find det kritiske punkt ved at sætte den første afledte lig med 0
critical_points = solve(f_prime, x)
```

```python
critical_points
```

```python
# Udregn funktionens værdi ved det kritiske punkt - hvis der er flere kritiske punkter
f_critical_values = [f.subs(x, cp) for cp in critical_points]
```

```python
# Hvis der kun er et kritisk punkt ville det være fint at skrive sådan.
f_critical_values2 = f.subs(x, critical_points[0])
```

```python
f_critical_values
```

```python
# Evaluer andet derivat i det kritiske punkt - med output som numerisk værdi (med evalf())
second_derivative_test = f_double_prime.evalf(subs={x: critical_points[0]})
```

```python
second_derivative_test
```

```python
# Definer x og y som numeriske værdier
x = np.linspace(-2, 2, 400)
y = x**3
```

```python
# plot funktion med markering ved kritisk punkt (saddelpunkt)

# Figurens størrelse
plt.figure(figsize=(8, 6))

#Plot af funktion
plt.plot(x, y, label='$f(x) = x^3$')

# Markering for kritisk punkt
plt.scatter(critical_points, f_critical_values, color='red')

#Titel
plt.title('Plot of $f(x) = x^3$')

# x- og y-label
plt.xlabel('x')
plt.ylabel('f(x)')

# Legend
plt.legend()

# Grid-system
plt.grid(True)

# Vis/afslut plot
plt.show()
```

Når vi finder det første derivat, som er $f'(x) = 3x^2$, ser vi, at det er 0, når $x = 0$. Men i dette tilfælde er det ikke et lokalt minimum eller maksimum, men snarere et saddelpunkt.

Grunden til, at $x=0$ ikke er et lokalt minimum eller maksimum, er fordi funktionens adfærd ikke ændrer sig fra stigende til faldende eller omvendt ved $x = 0$. For $x < 0$ er $f(x)$ negativ, og for $x > 0$ er $f(x)$ positiv. Så funktionen ændrer sig fra at være negativ til at være positiv ved $x=0$, men der er ikke nogen "top" eller "bund" ved $x=0$, som der ville være ved et lokalt maksimum eller minimum. I stedet passerer funktionen glidende igennem $x=0$, hvilket er karakteristisk for et saddelpunkt.

```python
# If/else betingelse, som testes:
if second_derivative_test < 0:
    print(f"x={critical_points[0]} er et maksimum.")
elif second_derivative_test > 0:
    print(f"x={critical_points[0]} er et minimum.")
else:
    print(f"x={critical_points[0]} er et saddelpunkt - eller testen er ikke afgørende.")
```

## Note om globale ekstrema
I mere komplekse funktioner med stort datagrundlag ($x$-værdier), kan det være en større opgave at definere globale kontra lokale ekstrema, men ofte kan man bruge simple NumPy metoder til at vurdere højeste eller laveste punkt for en funktion.

Funktionerne **np.argmin** og **np.argmax** fra NumPy-biblioteket returnerer indekset for den mindste/højeste værdi i en given array. Disse funktioner er særligt nyttige, når man skal finde positionen for det laveste/højeste punkt i en numerisk liste eller array, så du kan referere til det senere.

```python
import numpy as np

array = np.array([10, 20, 5, 40, 30])
min_index = np.argmin(array)

print("Indeks for minimumsværdi:", min_index)
```

```python
max_index = np.argmax(array)
print("Indeks for maksimumværdi:", max_index)
```
