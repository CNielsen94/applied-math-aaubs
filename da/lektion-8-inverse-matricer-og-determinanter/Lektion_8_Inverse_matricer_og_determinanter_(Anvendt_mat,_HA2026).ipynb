## Den inverse matrix

Nu ved vi lidt mere om identitetsmatricen og kan overveje hvordan vi annullerer
koefficientmatrixen $A$. Eller sagt på en anden måde: Vi vil prøve at transformere $A$ til identitetsmatricen $I$. Dette gøres ved at multiplicere den **inverse** matrix med matricen.



En invers matrix svarer til ideen om multiplikativ invers. Hvis vi, for eksempel, vil løse $x$ i ligningen $5x = 10$. For at gøre dette, har vi
brug for at multiplicere begge sider med den multiplikative inverse af $5$, som er $5^{-1}$ (eller $1/5$):

$5^{-1}*5x = 5^{-1} * 10$

Det omvendte af $5$ forvandler det til $1$ og efterlader os med løsningen: $x = 2$. For at løse for vektoren $\vec{x}$ i matrixligningen, er vi nødt til at gange begge sider med det inverse $A$:

$A^{-1}A\vec{x}=A^{-1}\vec{b}$

Dette kan simplificeres til $I\vec{x}=A^{-1}\vec{b}$, hvilket giver os formularen til at udregne løsningsvektoren:

$\vec{x} = A^{-1}\vec{b}$

Selvom vi bruger *inverse matrix* til at annullere bestemte i princippet på samme måde som når man bruger multiplikativ invers, er beregningen helt anderledes. Lad os forstå beregningen for det inverse af en 2 * 2 matrix.

Hvis $ A =
  \begin{bmatrix}
    a & b\\
    c & d \\
  \end{bmatrix}$
så $A^{-1} = \frac{1}{ad-bc}\begin{bmatrix}
    d & -b\\
    -c & a \\
  \end{bmatrix}$


Udtrykket $ad - bc$ er kendt som **determinanten** og skrives ofte som
eller som $det(A) = ad - bc$ eller som $|A| = ad - bc$. Fordi vi deler med determinanten ved beregning af den inverse matrice, er en 2 x 2 matrix kun inverterbar, hvis determinanten ikke er lig med 0.

I dette trin og det næste trin vil vi fokusere på
at finde den inverse matrix, når $A$ er en 2 x 2 matrix. Senere vil vi gennemgå, hvordan man beregner den inverse matrix til en højere dimensionel matrix (3 x 3 og derover).

Lad os implementere en invers matrix i Python før vi forsøger at løse matrixligningen.


**Øvelse**

* Opret en funktion ved navn **matrix_inverse_two()**, der accepterer en 2 x
2 matrix, som en NumPy ndarray, og returnerer en invers matrix. Denne funktion
skal først beregne matrixens determinant.
  * Hvis determinanten er lig med 0, skal funktionen rejse en fejl (`raise ValueError`).
  * Hvis determinanten ikke er lig med 0, skal denne funktion returnere invers matrix.
* Beregn det inverse af **matrix_a** ved hjælp af den funktion, du lige har skrevet og tildel resultatet til **inverse_a**.
* Multiplicer **inverse_a** med **matrix_a**, og tildel resultatet til **i_2**. Vis **i_2** ved hjælp af print() funktionen.



```python
# Hvad der foregår "under the hood" i NumPy
# Manuelt udført funktion til at udforme den inverse matrix (af en 2x2 matrix)
def matrix_inverse_two(M):
  det = M[0,0]*M[1,1] - M[0,1]*M[1,0]
  if det == 0:
    raise ValueError('Determinant = 0 - Det kan ikke lade sig gøre')
  else:
    return 1/det * np.array([[M[1,1],-M[0,1]],
                             [-M[1,0],M[0,0]]])
```

```python
# Generer matrice
matrix_a = np.array([[1,2],
                     [3,4]])
```

```python
# Manuel udførsel af den inverse matrix
inverse_a = matrix_inverse_two(matrix_a)
```

```python
# Vis inverse_a
inverse_a
```

```python
# Konvertering til identitetsmatrix
inverse_a @ matrix_a
```

```python
# Med NumPy (meget nemmere process)
np.linalg.inv(matrix_a) @ matrix_a
```

### Determinanten og den inverse matrix af fler-dimensionelle matricer

Før vi undersøger, hvordan man beregner den inverse matrix for fler-dimensionelle matricer, skal vi først dykke dybere ned i determinanten og introducere noget mere terminologi. Indtil videre har vi for det meste arbejdet med matricer, der indeholder det samme antal rækker og kolonner. Disse matricer er kendt som kvadratmatricer, og vi kan kun beregne determinanten og den inverse matrix for kvadratmatricer. Derudover kan vi kun beregne den inverse matrix af kvadratmatricer, når determinanten **ikke** er lig med 0.

For at finde determinanten for en fler-dimensionel kvadratmatrix skal vi bruge
den mere generelle form af determinanten. Den ser sådan ud:

$ det(
  \begin{bmatrix}
    a & b & c\\
    d & e & f \\
    g & h & i
  \end{bmatrix}) = a* det
  (\begin{bmatrix}
    e & f\\
    h & i \\
  \end{bmatrix})
  - b*det(
  \begin{bmatrix}
    d & f\\
    g & i \\
  \end{bmatrix}) + c* det
  (\begin{bmatrix}
    d & e\\
    g & h \\
  \end{bmatrix})$




Determinanten for et fler-dimensionelt system er forbundet med at nedbryde den fulde
matrix til mindre matricer. Først vælger vi en række eller kolonne (der vælges  tit den første række). For den første værdi i den valgte række "skjuler" vi de værdier i den række (2. og 3. værdi i rækken) og i den tilsvarende kolonne (2. og 3. værdi i kolonnen), benyt resten af elementerne som den mindre matrix, og gang den skalære værdi med determinanten for den mindre matrix. Vi gentager dette for de resterende værdier i første række.

Her er et konkret eksempel:

$ det(
  \begin{bmatrix}
    1 & 2 & 3\\
    4 & 5 & 6 \\
    7 & 8 & 9
  \end{bmatrix}) = 1* det
  (\begin{bmatrix}
    5 & 6\\
    8 & 9 \\
  \end{bmatrix})
  - 2*det(
  \begin{bmatrix}
    4 & 6\\
    7 & 9 \\
  \end{bmatrix}) + 3* det
  (\begin{bmatrix}
    4 & 5\\
    7 & 8 \\
  \end{bmatrix})$

For at beregne determinanten i NumPy bruger vi funktionen **numpy.linalg.det()**. Mere om denne funktion kan findes i [dokumentationen](https://numpy.org/doc/stable/reference/generated/numpy.linalg.det.html) hvor i kan lære mere om, hvordan i bruger denne funktion.

For at beregne en invers matrix for en 3 * 3 eller større matrix, skal vi også arbejde med den mere generelle form af matrixens inverse ligning. Svarende til determinanten for fler-dimensionelle matricer fungerer den inverse matrix ved at generere mindre matricer, der er afhængige af positionen i matrixen. Her er et diagram, der beskriver den inverse matrix for en 3 * 3 matrix:




$
\begin{bmatrix}
    a & b & c\\
    d & e & f \\
    g & h & i
\end{bmatrix} ^{-1} =
\frac{1}{det(\begin{bmatrix}
    a & b & c\\
    d & e & f \\
    g & h & i
\end{bmatrix})}
\begin{bmatrix}
    det(\begin{bmatrix}
    a & - & -\\
    - & e & f \\
    - & h & i \end{bmatrix})  
  & ...
  & det(\begin{bmatrix}
    - & - & c\\
    d & e & - \\
    g & h & - \end{bmatrix}) \\
... & ... & ... \\
det(\begin{bmatrix}
    - & b & c\\
    - & e & f \\
    g & - & - \end{bmatrix})  
  & ...
  & det(\begin{bmatrix}
    a & b & -\\
    d & e & - \\
    - & - & i \end{bmatrix}) \\
\end{bmatrix}
$

Selvom det er nyttigt at vide, hvordan man beregner det inverse på denne måde til fler-dimensionelle matricer, er mængden af omhyggelig aritmetik, du skal udføre i hånden, stor. Heldigvis, kan funktionen **numpy.linalg.inv()** arbejde med enhver n-dimensionel kvadratmatrix.


En matrix er en rektangulær tabel af tal, symboler eller udtryk, arrangeret i rækker og kolonner. Matricer bruges til at repræsentere lineære transformationer, systemer af lineære ligninger, og mange andre koncepter inden for lineær algebra og matematik generelt.

Den inverse af en matrix A, som normalt betegnes som A^-1, er en speciel form for matrix. Hvis du multiplicerer en matrix med dens inverse, får du identitetsmatricen (som er matematikkens ækvivalent til tallet 1 inden for matrixmultiplikation).

For eksempel, hvis vi har en matrix A:

$A = [[a, b],
[c, d]]$

dens inverse er:

$A^-1 = 1/(ad-bc) * [[d, -b],
[-c, a]]$

Hvor "1/(ad-bc)" er den reciprokke (eller "omvendte") af determinanten af matrix A.

**Bemærk**, at ikke alle matricer har en inverse. Kun kvadratiske matricer (dvs. matricer med lige mange rækker og kolonner) kan have en inverse, og selv da har kun de matricer, hvis determinant ikke er nul, en inverse. Disse matricer kaldes invertible eller non-singulære.

Den inverse matrix spiller en central rolle i mange anvendelser af lineær algebra, herunder løsning af systemer af lineære ligninger, beregning af matrixpotenser og mange andre anvendelser.

## Videre information om determinanten

En determinant er en speciel værdi, der kan beregnes fra elementerne i en kvadratisk matrix. Determinanten har en række egenskaber og mange anvendelser inden for matematik. Det mest almindelige sted at støde på determinanter er inden for systemer af lineære ligninger, da det hjælper med at finde, om systemet har en unik løsning.

Lad os se på en 2x2 matrix:

A = [[a, b],
[c, d]]

Determinanten af denne matrix, ofte skrevet som |A| eller det(A), beregnes ved:

|A| = ad - bc

Determinanten har en række interessante egenskaber. For eksempel, hvis determinanten af en matrix er 0, siger vi, at matrixen er "singulær" eller "degenereret", hvilket betyder, at den ikke har en inverse. Dette hænger sammen med den idé, vi talte om før - en matrix har kun en inverse, hvis determinanten ikke er 0.

I større matricer er beregning af determinanten lidt mere kompleks, men følger stadig mønstre baseret på elementerne i matrixen.

Generelt er determinanten vigtig, fordi den fortæller os meget om matrixen, inklusive om vi kan finde dens inverse. Dette er kritisk for mange anvendelser af lineær algebra, herunder løsning af systemer af lineære ligninger.

## Input-output modeller (Leontief modellen)

Udtrykket $(I - M)^{-1}$ refererer til den inverse matrix af $(I−M)$, hvor $I$ er identitetsmatricen og $M$ er inputmatricen for økonomien. Dette er en central del af den såkaldte Leontief Input-Output Model, der er en metode til at beskrive, hvordan forskellige sektorer i en økonomi interagerer med hinanden.

I praksis bruges  $(I - M)^{-1}$ til at beregne den samlede output $(X)$ fra hver sektor, der er nødvendig for at opfylde en given endelig efterspørgsel $(D)$. Matematisk er dette givet ved:

$X=(I−M)^{−1} ×D$

Her repræsenterer $D$ den endelige efterspørgsel fra hver sektor, og $X$ er den samlede output fra hver sektor, der er nødvendig for at opfylde denne efterspørgsel.

Hvorfor er det vigtigt?

* Det tager højde for den indbyrdes afhængighed mellem forskellige sektorer. For eksempel, hvis sektor A kræver input fra sektor B til sin produktion, og sektor B i sin tur kræver input fra sektor A, skal begge disse relationer medtages i analysen.

* Det giver en måde at beregne, hvordan ændringer i efterspørgslen i én sektor vil påvirke output og efterspørgsel i andre sektorer.

* Det er særligt nyttigt for politikere, økonomer og erhvervsledere til at forstå, hvordan ændringer i én sektor kan have en kaskadeeffekt på hele økonomien.


#### Eksempel med Leontief modellen

En økonomi er baseret på tre sektorer: handel (H), transport (T) og sundhed (S). Produktion af en kornes værdi af handel kræver en input af 0.25dkk fra handelssektoren og 0.15dkk fra transportsektoren. Produktion af en krones værdi af transport kræver en input af 0.05dkk fra transportsektoren og 0.35dkk fra sundhedssektoren. Produktion af en krones værdi af sundhed kræver en input af 0.20dkk fra handelssektoren, 0.05dkk fra transportsektoren, og 0.25dkk fra sundhedssektoren.

* Find inputmatricen $M$ for denne økonomi
* Udregn $(I - M)^{-1}$ ved hjælp af Python
* Find output fra hver sektor, der er nødvendig for at opfylde en endelig efterspørgsel på 15 milliarder dkk for handel, 5 milliarder dkk for transport, og 25 milliarder dkk for sundhed.


```python
# Definer inputmatricen M
# Kolonne j = hvad 1 krones produktion i sektor j kræver fra hver sektor (rækkerne)
#     produceret i:  H     T     S
M = np.array([
    [0.25, 0.00, 0.20],   # input fra handel (H)
    [0.15, 0.05, 0.05],   # input fra transport (T)
    [0.00, 0.35, 0.25]    # input fra sundhed (S)
])
```

```python
# Udregn (I - M)^{-1}
I = np.identity(3)  # 3x3 Identity matrix
IM_inv = np.linalg.inv(I - M)
```

```python
# Udregn ouput for hver sektor
D = np.array([15, 5, 25])  # vektor for endelig efterspørgsel
X = IM_inv @ D
```

```python
print("Input Matrix M:")
print(M)
print("\n (I - M)^{-1}:")
print(IM_inv)
print("\n Output fra hver sektor, for at efterspørgsel mødes:")
print(X)
```

## Eksempel med lineært system

$A\vec{x} = \vec{b}$

$$
\text{Ligningssystemet er givet ved:}
\begin{aligned}
\begin{smallmatrix} \textrm{A}\\\begin{bmatrix}
2 & 1 & 4 & 1 \\
3 & 4 & -1 & 2 \\
1 & -1 & 3 & 2 \\
2 & 7 & 1 & 5 \\
\end{bmatrix}\end{smallmatrix}
*
\begin{smallmatrix} \textrm{x}\\\begin{bmatrix}
x_1 \\
x_2 \\
x_3 \\
x_4 \\
\end{bmatrix}\end{smallmatrix}
&=
\begin{smallmatrix} \textrm{B}\\\begin{bmatrix}
15 \\
10 \\
13 \\
23 \\
\end{bmatrix}\end{smallmatrix}
\end{aligned}
$$

$$
\text{Det kan også udtrykkes som enkelte ligninger: }
\begin{aligned}
2x_1 + 1x_2 + 4x_3 + 1x_4 &= 15, \\
3x_1 + 4x_2 - 1x_3 + 2x_4 &= 10, \\
1x_1 - 1x_2 + 3x_3 + 2x_4 &= 13, \\
2x_1 + 7x_2 + 1x_3 + 5x_4 &= 23.
\end{aligned}
$$


```python
# Definerer en 4x4 matrix A og en kolonnevektor B
A = np.array([[2, 1, 4, 1],
              [3, 4, -1, 2],
              [1, -1, 3, 2],
              [2, 7, 1, 5]])

B = np.array([15, 10, 13, 23])
```

```python
# Beregner determinanten for A
det_A = np.linalg.det(A)
```

```python
# Kontrollerer om determinanten er forskellig fra 0 (hvis den er 0, har matrixen ingen invers)
if det_A == 0:
    print("Matrixen har ingen invers.")
else:
    # Beregner den inverse matrix af A
    A_inv = np.linalg.inv(A)

    # Løser for X ved at multiplicere A_inv med B
    X = A_inv @ B

    print("Løsningssættet X er:", X)
```

* Hvis determinanten af en 2x2 matrix, eller $ad-bc$ er lig med 0, er matricen **ikke** inverterbar. Man kan udelukkende udregne determinanten og den inverse matrice for matricer der har den samme mængde rækker og kolonner.
* For at udregne determinanten for større matricer (3x3 og opefter) kan vi opdele den fulde matrix i mindre matricer.
* Identitetsmatricen indeholder værdien $1$ på diagonalen, og værdien $0$ på de resterende positioner. Identitetsmatricen er ofte repræsenteret ved $I_n$ hvor $n$ er antallet af rækker og kolonner.
* når vi multiplicerer med en vilkårlig vektor indeholdende 2 elementer, vil den resulterende vektor matche originale vektor fuldstændig.  
* For at transformere $A$ til en identitetsmatrice $I$ i $A\vec{x} = \overrightarrow{b}$, multiplicerer vi begge sider med det inverse af matrix $A$.
