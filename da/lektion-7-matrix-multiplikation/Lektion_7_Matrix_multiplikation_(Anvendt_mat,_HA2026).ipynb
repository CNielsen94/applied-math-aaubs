Den følgende matrixligning er et eksempel på matrix-vektor-multiplikation. Når vi multiplicerer en matrix med en vektor, kombinerer vi i det væsentlige hver række af matrixen med kolonnevektoren.

$$
Ac =
  \begin{smallmatrix} \textrm{A}\\\begin{bmatrix}
    0.7 & 3 & 9 \\
    1.7 & 2 & 9 \\
    0.7 & 9 & 2 \\
  \end{bmatrix}\end{smallmatrix}
*
 \begin{smallmatrix} \textrm{c}\\\begin{bmatrix}
    1 \\
    2 \\
    1 \\
  \end{bmatrix}\end{smallmatrix}
=
\begin{smallmatrix} \textrm{A * c}\\\begin{bmatrix}
    (0.7*1) &+& (3*2) &+& (9*1) \\
    (1.7*1) &+& (2*2) &+& (9*1) \\
    (0.7*1) &+& (9*2) &+& (2*1) \\
  \end{bmatrix}\end{smallmatrix}
=
  \begin{smallmatrix} \textrm{Ac}\\\begin{bmatrix}
    15.7 \\
    14.7 \\
    20.7 \\
  \end{bmatrix}\end{smallmatrix}
$$

For at multiplicere en matrix med en vektor skal antallet af kolonner i matrixen matche antallet af rækker i vektoren.

Gyldige produkter:
$$
  \begin{bmatrix}
    0.7 & 3 & 9 \\
    1.7 & 2 & 9 \\
    0.7 & 9 & 2
  \end{bmatrix}
·%
  \begin{bmatrix}
    1 \\
    2 \\
    1
  \end{bmatrix}
 %
 \hspace{1cm}
  \begin{bmatrix}
    0.7 & 3 & 9 \\
    1.7 & 2 & 9 \\
  \end{bmatrix}
·%
  \begin{bmatrix}
    1 \\
    2 \\
    1
  \end{bmatrix}
$$

Nej! 🚫
$$
  \begin{bmatrix}
    0.7 & 3 & 9 \\
    1.7 & 2 & 9 \\
  \end{bmatrix}
·
  \begin{bmatrix}
    1 \\
    2 \\
  \end{bmatrix}
$$

Ligesom med 2 vektorer kan vi bruge `numpy.dot()` eller `@` til at beregne produktet.

Som vi ved, består en matrix af søjlevektorer. Derfor kan vi bruge de samme koncepter, vi lærte om matrix-vektor-multiplikation, til at multiplicere matricer med hinanden. I matrix-vektor-multiplikation udførte vi et prikprodukt mellem hver række af den første matrix og søjlevektoren. I matrixmultiplikation udvider vi dette til at udføre et prikprodukt mellem hver række af den første matrix og hver kolonne af den anden matrix.

$$
AB =
  \begin{smallmatrix} \textrm{A}\\\begin{bmatrix}
    0.7 & 3 \\
    1.7 & 2 \\    
  \end{bmatrix}\end{smallmatrix}
·
 \begin{smallmatrix} \textrm{B}\\\begin{bmatrix}
    1 & 2\\
    2 & 3\\
  \end{bmatrix}\end{smallmatrix}
=
\begin{smallmatrix} \textrm{A * B}\\\begin{bmatrix}
    (0.7*1+3*2) & (0.7*2+3*3) \\
    (1.7*1+2*2) & (1.7*2+2*3) \\
  \end{bmatrix}\end{smallmatrix}
=
  \begin{smallmatrix} \textrm{AB}\\\begin{bmatrix}
    6.7 & 10.4 \\
    5.7 & 9.4 \\
  \end{bmatrix}\end{smallmatrix}
$$

Ligesom med matrix-vektor-multiplikation skal kolonnerne i den første matrix matche antallet af rækker i den anden matrix.

```python
A = np.array([[0.7, 3],
              [1.7 , 2]])

B = np.array([[1, 2],
              [2 , 3]])

A @ B
```

BEMÆRK: Rækkefølgen er vigtig!

$$
BA =
  \begin{smallmatrix} \textrm{B}\\\begin{bmatrix}
    1 & 2 \\
    2 & 3 \\    
  \end{bmatrix}\end{smallmatrix}
·
 \begin{smallmatrix} \textrm{A}\\\begin{bmatrix}
    0.7 & 3\\
    1.7 & 2\\
  \end{bmatrix}\end{smallmatrix}
=
\begin{smallmatrix} \textrm{B * A}\\\begin{bmatrix}
    (1*0.7+2*1.7) & (1*3+2*2) \\
    (2*0.7+3*1.7) & (2*3+3*2) \\
  \end{bmatrix}\end{smallmatrix}
=
  \begin{smallmatrix} \textrm{BA}\\\begin{bmatrix}
    4.1 & 7 \\
    6.5 & 12 \\
  \end{bmatrix}\end{smallmatrix}
$$

For at multiplicere matricer i NumPy skal du bruge den samme `numpy.dot()` funktion eller `@` operator, som vi brugte tidligere.

Matrix-transponeringen ombytter rækker og kolonner i en matrix. Du kan tænke på transponeringen som en rotation. Inden for dataanalyse, statistik og finans bruges datatabeller af forskellige dimensioner ofte. På grund af kravene til matrixmultiplikation skal vi nogle gange transponere en matrix, så vi kan multiplicere matricer med hinanden, der ikke har det samme antal rækker og kolonner.

En matrix-transponering vil typisk se sådan ud:

Original matrix:
$$
  \begin{bmatrix}
    1 & 2\\
    3 & 4 \\
    5 & 6 \\
  \end{bmatrix}
$$

Efter transponering:

$$
  \begin{bmatrix}
    1 & 3 & 5\\
    2 & 4 & 6\\
  \end{bmatrix}
$$

Matematisk bruger vi notationen $A^T$ til at angive transponeringsoperationen.

$A^T + B^T = C$

Transponeringen har nogle interessante regler, der er ret intuitive. For eksempel, når vi tager transponeringen af summen af to matricer, kan vi distribuere transponeringen til hver matrix:

$(A + B)^T = A^T + B^T$

Det er dog lidt mere kontraintuitivt, når man tager transponeringen af produktet af to matricer:

$(AB)^T = B^T A^T$

Lad os se dette med NumPy: Vi kan enten bruge `np.transpose()` eller simpelthen skrive `.T` efter en matrix.

##### **Nøglekoncepter**
*   Mange operationer, der kan udføres på vektorer, kan også udføres på matricer. Vi kan udføre følgende operationer med matricer:
    *   Lægge sammen og trække fra matricer, der har det samme antal rækker og kolonner.
    *   Multiplicere en matrix med en skalarværdi.
    *   Multiplicere en matrix med en vektor og andre matricer. For at multiplicere med en vektor eller en anden matrix skal antallet af kolonner matche antallet af rækker i vektoren eller matrixen.

##### NumPys random-funktion

**Random seed**

Du kan bruge tilfældige tal fra NumPy og stadig få de samme resultater hver gang, hvis du bruger

`np.random.seed(42)`

```python
# Bonus: tilfældige tal i NumPy - med np.random.seed() bliver "tilfældige" tal reproducerbare
np.random.seed(42)
np.random.rand(2,2)
```

```python
# Samme seed giver PRÆCIS de samme tal - prøv at sammenligne med cellen ovenfor
np.random.seed(42)
np.random.rand(2,2)
```

**Random.rand**

`np.random.` bruges til tilfældig sampling og anvendes ofte inden for matematik, statistik, finans og forskellige videnskabelige applikationer.

For eksempel bruges `np.random.rand()` til at generere tilfældige værdier baseret på forskellige parametre. Følgende er eksempler på, hvordan `np.random` kan bruges i forskellige situationer, men mere information kan findes [her](https://numpy.org/doc/stable/reference/random/index.html).

```python
# Generer en enkelt tilfældig værdi
tilfældig_værdi = np.random.rand()
print(f"Enkelt tilfældig værdi: {tilfældig_værdi}")

print("\n")

# Generer et array med 5 tilfældige værdier
tilfældigt_array = np.random.rand(5)
print(f"Array med 5 tilfældige værdier: {tilfældigt_array}")

print("\n")

# Generer en 2x2 matrix med tilfældige værdier
tilfældig_matrix = np.random.rand(2, 2)
print(f"2x2 matrix med tilfældige værdier: \n{tilfældig_matrix}")
```

```python
np.random.seed(42)

# Generer et tilfældigt heltal mellem 1 og 10
tilfældigt_heltal = np.random.randint(1, 11)
print(f"Tilfældigt heltal mellem 1 og 10: {tilfældigt_heltal}")

print("\n")

# Generer et array med 5 tilfældige heltal mellem 1 og 10
tilfældigt_heltals_array = np.random.randint(1, 11, 5)
print(f"Array med 5 tilfældige heltal mellem 1 og 10: {tilfældigt_heltals_array}")

print("\n")

# Generer en 3x3 matrix med tilfældige heltal mellem 1 og 10
tilfældig_heltals_matrix = np.random.randint(1, 11, (3, 3))
print(f"3x3 matrix med tilfældige heltal mellem 1 og 10: \n{tilfældig_heltals_matrix}")
```

```python
# Tilfældig blanding af et array
arr = np.array([1, 2, 3, 4, 5])

np.random.shuffle(arr)


print(f"Blandet array: {arr}")
print("\n")

# Generer en tilfældig permutation
perm = np.random.permutation(5)
print(f"Tilfældig permutation: {perm}")
print("\n")

tilfældig_perm = np.random.permutation(arr)
print(f"Tilfældig permutation af array: {tilfældig_perm}")
```

`np.random.permutation` har ikke indbygget funktionalitet til at angive et startpunkt eller antal elementer direkte som parametre. Funktionen tager enten et heltal `x` og returnerer en tilfældig permutation af `np.arange(x)` (som starter fra 0 og går op til `x-1`), eller den tager et eksisterende array og returnerer en tilfældig permutation af dette array.

```python
#np.random.seed(42)

# Opret en liste af valgmuligheder
valgmuligheder = ['apple', 'banana', 'lemon', 'pineapple', 'pear', 'orange', 'mandarin']

# Vælg et tilfældigt element fra listen
tilfældigt_valg = np.random.choice(valgmuligheder)

print(f"Tilfældigt valgt frugt: {tilfældigt_valg}")
```

```python
# Vælg fra valgmuligheder med forskellige sandsynligheder
tilfældigt_valg_med_sandsynlighed = np.random.choice(valgmuligheder, p=[0.1, 0.2, 0.3, 0.1, 0.1, 0.1, 0.1])
print(f"Valgt frugt med angivet sandsynlighed: {tilfældigt_valg_med_sandsynlighed}")
```

Husk, at parameteren `p` henviser til sandsynlighed. Værdierne i `p` skal både matche antallet af observationer i den givne liste og summe til 1. Generelt er det en god idé at forstå hyperparametrene og funktionaliteten af de forskellige pakker, du bruger i Python. Det er ofte muligt at opnå betydelig optimering ved at bruge parametre korrekt og rettidigt.
